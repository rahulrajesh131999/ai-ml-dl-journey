## PCA in one sentence

**PCA finds new axes (directions) that capture the most variation in the data, then keeps only the most important ones.**

---

## 1. Why do PCA?

Suppose your data has **100 features**.

Many of them contain **similar information** (are correlated).

Instead of keeping all 100, PCA creates a smaller set (say 10) that preserves most of the information.

---

## 2. What are the "axes"?

They are **new directions**, not the original features.

Imagine rotating the coordinate system until it lines up with how the data is spread.

```
Original axes:        PCA axes:

y                     PC2
|                    /
|                 /
|              /
|________x    /________ PC1
```

PC1 = direction of maximum spread.

PC2 = next largest spread (perpendicular to PC1).

---

## 3. Why maximum spread?

More spread (variance) means **more information**.

Little spread means the data barely changes in that direction.

So we keep directions with the largest variance.

---

## 4. What does "orthogonal" mean?

The new axes are **90° apart**.

This means they are **uncorrelated**.

No duplicated information.

---

## 5. What is the covariance matrix?

It summarizes

* variance of each feature
* correlation between every pair of features

Diagonal → variances

Off-diagonal → covariances

---

## 6. Why eigenvectors?

The covariance matrix tells us how the data is spread.

Its

* **eigenvectors = the new axes (principal components)**
* **eigenvalues = how much variance each axis captures**

This is the key fact of PCA.

---

## 7. Sort by eigenvalues

Example:

| PC  | Eigenvalue |
| --- | ---------: |
| PC1 |         18 |
| PC2 |          7 |
| PC3 |          2 |
| PC4 |        0.5 |

Keep the largest ones because they explain the most variation.

---

## 8. What does "top-k" mean?

If

* original dimension = 100
* desired dimension = 10

keep

PC1 ... PC10

discard the remaining 90.

---

## 9. Final step

Project every data point onto these new axes.

Instead of

```
(x1,x2,...,x100)
```

you now have

```
(PC1,PC2,...,PC10)
```

Same data, fewer dimensions.

---

## 10. The entire algorithm

```
Data
   ↓
Compute covariance matrix
   ↓
Find eigenvectors & eigenvalues
   ↓
Sort by eigenvalue (largest first)
   ↓
Keep top-k eigenvectors
   ↓
Project data onto them
   ↓
Reduced-dimensional data
```

---

# The only sentence you need to remember

> **PCA rotates the feature space to find directions of maximum variance (principal components), ranks them by their variance (eigenvalues), and projects the data onto the top (k) directions to reduce dimensionality while preserving as much information as possible.**

Everything in that passage is just explaining **why** the covariance matrix, eigenvectors, and eigenvalues let us find those "best directions."

Yes, exactly.

When you write:

```python
pca = PCA(n_components=2)
```

you are asking PCA to **reduce the data to 2 dimensions** by creating **2 new features** (called principal components).

Think of it like this:

* Suppose your original dataset has **10 features**:

  ```
  X = [x1, x2, x3, ..., x10]
  ```
* PCA combines these 10 features into **2 new features**:

  ```
  PC1, PC2
  ```
* So each data point changes from:

  ```
  [x1, x2, ..., x10]   →   [PC1, PC2]
  ```

### Is it compression?

Yes. PCA is a form of **dimensionality reduction**, which is essentially **compressing the feature space**.

The goal is to keep as much information (variance) as possible while using fewer dimensions.

For example:

| Original                   | After PCA                 |
| -------------------------- | ------------------------- |
| 1000 samples × 50 features | 1000 samples × 2 features |

Instead of storing 50 values for each sample, you now store only 2 values.

### How does PCA do this?

The new features are **not selected from the original features**. Instead, they are **linear combinations** of them.

For example:

```
PC1 = 0.5x1 + 0.3x2 - 0.2x3 + ... + 0.1x10
PC2 = -0.1x1 + 0.7x2 + 0.4x3 + ... - 0.3x10
```

These combinations are chosen so that:

* **PC1** captures the maximum possible variance in the data.
* **PC2** captures the maximum remaining variance while being orthogonal (independent) to PC1.

### An analogy

Imagine taking a photo of a 3D object.

* The object is **3D** (or even higher-dimensional).
* The photo is **2D**.
* If you choose the camera angle well, the photo still preserves most of the important structure.

PCA does something similar: it finds the **best 2D "view"** of your higher-dimensional data, preserving as much information (variance) as possible.

So yes—**`n_components=2` means compressing the original features into a 2-dimensional representation using PCA**, with the aim of retaining as much of the original information as possible.
