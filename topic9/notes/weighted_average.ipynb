Of course! Think of a **weighted moving average (WMA)** as a moving average that **gives more importance to recent data**.

### 1. Regular Moving Average

Suppose sales for the last 3 days are:

| Day   | Sales |
| ----- | ----: |
| Day 1 |    10 |
| Day 2 |    20 |
| Day 3 |    30 |

A **3-day simple moving average** gives each day the same importance:

[
\frac{10+20+30}{3}=20
]

Each observation has a weight of **1/3 = 0.333**.

---

### 2. Weighted Moving Average

Now suppose we believe **recent sales are more important**.

We choose weights:

* Most recent day: **0.5**
* Previous day: **0.3**
* Oldest day: **0.2**

Notice:

[
0.5+0.3+0.2=1
]

The weights must always add up to **1**.

Now calculate:

[
(30\times0.5)+(20\times0.3)+(10\times0.2)
]

Step by step:

* (30\times0.5=15)
* (20\times0.3=6)
* (10\times0.2=2)

Total:

[
15+6+2=23
]

So the **weighted moving average = 23**.

Notice that:

* Simple Moving Average = **20**
* Weighted Moving Average = **23**

Since the latest value (30) is larger, giving it a higher weight increases the average.

---

### 3. Understanding the Formula

Your formula is

[
\hat{y}*{t}=\sum*{n=1}^{k}\omega_n y_{t+1-n}
]

Here's what each symbol means:

* (\hat{y}_t) = forecast for time (t)
* (k) = number of past observations used
* (\omega_n) = weight for each observation
* (y_{t+1-n}) = previous observations

For **3 observations**, it becomes:

[
\hat{y}_t
=========

0.5y_t
+
0.3y_{t-1}
+
0.2y_{t-2}
]

where

* (y_t) = most recent value
* (y_{t-1}) = one period ago
* (y_{t-2}) = two periods ago

---

### 4. Another Example

Suppose monthly sales are:

| Month    | Sales |
| -------- | ----: |
| January  |    80 |
| February |    90 |
| March    |   100 |

Weights:

* March = 0.6
* February = 0.3
* January = 0.1

Forecast:

[
100(0.6)+90(0.3)+80(0.1)
]

Calculate:

* (100\times0.6=60)
* (90\times0.3=27)
* (80\times0.1=8)

Forecast:

[
60+27+8=95
]

So the weighted moving average is **95**.

---

### 5. Why Use a Weighted Moving Average?

Imagine you're predicting tomorrow's weather.

* The temperature **yesterday** is more useful than the temperature **three weeks ago**.
* Therefore, yesterday should get a **larger weight**.

Similarly, in sales forecasting:

* Recent sales better reflect current demand.
* Older sales are less relevant.

That's why we assign **larger weights to recent observations**.

### Summary

| Simple Moving Average              | Weighted Moving Average                |
| ---------------------------------- | -------------------------------------- |
| All observations have equal weight | Recent observations get larger weights |
| Example weights: 1/3, 1/3, 1/3     | Example weights: 0.5, 0.3, 0.2         |
| Average of values                  | Weighted sum of values                 |
| Slower to react to changes         | Faster to react to recent changes      |

**Easy way to remember:**

* **Simple Moving Average:** "Everyone gets an equal vote."
* **Weighted Moving Average:** "Recent data gets a louder vote."
