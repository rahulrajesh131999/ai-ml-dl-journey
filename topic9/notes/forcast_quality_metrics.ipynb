
# [R squared](http://scikit-learn.org/stable/modules/model_evaluation.html#r2-score-the-coefficient-of-determination):
## R squared: coefficient of determination
### Step 1: Imagine you know nothing

Suppose your friend asks:

  
> "Predict tomorrow's house price."

  

You know nothing except that the average house price is **$300,000**.

  

So for every house, you predict:

  

> **$300,000**

  

This is called the **mean model**.

  

It isn't smart, but it's your baseline.

  

---

  

## Step 2: How bad is this prediction?

  

For every house, compute

  

$$\text{Actual Price} - \text{Mean Price}$$

  

Square these differences and add them up.

  

This gives

  

$$SS_{tot}$$

  

Think of it as

  

> **How much variation exists in the data if I only use the average?**

  

StatQuest calls this the **mean variance** (more precisely, the total variation around the mean).

  

Suppose

  

$$SS_{tot}=100.$$

  

---

  

## Step 3: Now build a regression line

  

Instead of always predicting the mean, you fit a regression line.

  

Now each house gets its own prediction.

  

Some predictions are close.

Some are not.

  

Now compute

  

$$\text{Actual}-\text{Predicted}$$

  

Square them and add them.

  

This is

  

$$SS_{res}.$$

  

Suppose

  

$$SS_{res}=30.$$

  

This is the variation that **the regression failed to explain**.

  

---

  

## Step 4: How much variation did the line explain?

  

Originally there were

  

$$100$$

  

units of variation.

  

After using the regression line, only

  

$$30$$

  

units remain unexplained.

  

So the line explained

  

$$100-30=70$$

  

units.

  

StatQuest calls this the **line variance** or **explained variance**.

  

---

  

## Step 5: What percentage did the line explain?

  

This is exactly what $R^2$ measures.

  

> Out of the total variation, how much did the line explain?

  

So

  

$$R^2 = \frac{\text{Explained Variation}}{\text{Total Variation}} = \frac{70}{100} = 0.7$$

  

This means

  

> **The regression line explained 70% of the variation in the data.**

  

This is exactly StatQuest's explanation.

  

---

  

## Step 6: Now look at the textbook equation

  

The textbook writes

  

$$R^2=1-\frac{SS_{res}}{SS_{tot}}$$

  

Instead of saying

  

> Explained variation = 70

  

it says

  

> Unexplained variation = 30

  

Since

  

$$\text{Explained} = \text{Total} - \text{Unexplained},$$

  

we get

  

$$\frac{\text{Explained}}{\text{Total}} = \frac{\text{Total}-\text{Unexplained}}{\text{Total}}$$

  

Split the fraction:

  

$$\frac{\text{Total}}{\text{Total}} - \frac{\text{Unexplained}}{\text{Total}}$$

  

which becomes

  

$$1-\frac{SS_{res}}{SS_{tot}}.$$

  

So the textbook isn't introducing a new idea—it is just expressing the same concept differently.

  

---

  

## Here's the picture StatQuest has in mind

  

```

Total Variation (SS_tot)  ======================== 100

  

Regression explains       ================== 70

  

Still unexplained         ====== 30

```

  

So

  

$$R^2 = \frac{70}{100} = 0.70.$$

  

Or equivalently,

  

$$R^2 = 1-\frac{30}{100} = 0.70.$$

  

Both formulas describe the **same picture**.

  

---

  

### The key idea to remember

  

Whenever you see

  

$$R^2 = 1-\frac{SS_{res}}{SS_{tot}},$$

  

read it in English as:

  

> **Start with all the variation in the data (100%). Subtract the fraction that the regression could not explain. Whatever remains is the fraction that the regression did explain.**

  

That's exactly the intuition Josh Starmer uses in StatQuest. The textbook formula is simply the algebraic version of that explanation.


# [Mean Absolute Error](http://scikit-learn.org/stable/modules/model_evaluation.html#mean-absolute-error)

Let's explain **MAE** the same way Josh Starmer would: start with the intuition, then show where the formula comes from.

---

# Step 1: Imagine you're forecasting sales

Suppose your model predicts the number of ice creams sold each day.

|Day|Actual Sales ($y$)|Predicted Sales ($\hat y$)|
|---|--:|--:|
|1|100|90|
|2|80|95|
|3|120|125|
|4|90|85|

Your question is:

> **On average, how far are my predictions from the actual values?**

Notice we're **not** asking whether the prediction is higher or lower.

We're only asking:

> **How wrong am I?**

---

# Step 2: Calculate the error

For each prediction,

$$\text{Error} = y_i-\hat y_i$$

|Day|Error|
|---|---|
|1|100−90 = **10**|
|2|80−95 = **−15**|
|3|120−125 = **−5**|
|4|90−85 = **5**|

Some errors are positive.

Some are negative.

---

# Step 3: Ignore the direction

Imagine you're throwing darts at a target.

If one dart lands **10 cm left** and another lands **10 cm right**, are you perfectly accurate?

No!

You're still **10 cm away** from the center in both cases.

That's why we take the **absolute value**.

|Error|Absolute Error|
|---|---|
|10|10|
|−15|15|
|−5|5|
|5|5|

Now every error simply means

> **How far away was the prediction?**

---

# Step 4: Find the average mistake

Now average all those distances.

$$\frac{10+15+5+5}{4}=8.75$$

So your model is, **on average**, off by

> **8.75 ice creams per day.**

That's MAE.

---

# Step 5: The formula

$$MAE=\frac{\sum_{i=1}^{n}|y_i-\hat y_i|}{n}$$

Read it exactly like this:

- $y_i$ = Actual value
- $\hat y_i$ = Predicted value
- $y_i-\hat y_i$ = Prediction error
- $|y_i-\hat y_i|$ = Distance between prediction and reality
- Sum all those distances
- Divide by the number of observations

In plain English:

> **MAE is the average distance between your predictions and the actual values.**

---

# Why do we use the absolute value?

Suppose your errors are

$$10,\ -10,\ 20,\ -20$$

If you simply averaged them,

$$\frac{10-10+20-20}{4}=0$$

It would look like your model made **no mistakes**!

But that's obviously wrong.

Your predictions were actually quite far off.

Taking the absolute value prevents positive and negative errors from canceling each other.

---

# Why is MAE easy to understand?

Suppose you're predicting house prices.

MAE = **$5,000**

That means

> **On average, every prediction is off by about $5,000.**

Suppose you're predicting temperatures.

MAE = **2°C**

That means

> **On average, each prediction misses by about 2°C.**

The units stay exactly the same as the original data, which makes MAE very intuitive.

---

# Compare it to $R^2$

Notice how different the questions are:

- **$R^2$** asks:
    
    > _How much of the variation did my model explain?_ (Expressed as a proportion or percentage.)
    
- **MAE** asks:
    
    > _On average, how far away are my predictions from the true values?_ (Expressed in the same units as the data.)
    

So if you're forecasting daily sales and get:

- $R^2 = 0.90$: your model explains **90% of the variation** in sales.
- $MAE = 8.75$: your predictions are **off by about 9 sales per day on average**.

That's why MAE is one of the most interpretable forecasting metrics—it answers the simple question: **"How wrong am I, on average?"**


# [Median Absolute Error](http://scikit-learn.org/stable/modules/model_evaluation.html#median-absolute-error)

Perfect. Now you've learned **R²** and **MAE**, and **MedAE** is actually very easy because it differs from MAE by just **one word**:

> **Average → Median**

Let's build the intuition exactly the same way.

---

# Step 1: Predict some values

Suppose your model predicts daily sales.

|Day|Actual|Predicted|
|---|---|---|
|1|100|90|
|2|80|95|
|3|120|125|
|4|90|85|
|5|150|145|

---

# Step 2: Calculate the errors

$$y_i-\hat y_i$$

|Error|
|---|
|10|
|-15|
|-5|
|5|
|5|

---

# Step 3: Take absolute values

Just like MAE, we don't care whether we predicted too high or too low.

We only care **how far away** we were.

|Absolute Errors|
|---|
|10|
|15|
|5|
|5|
|5|

---

# Step 4: Here's where MedAE becomes different

For **MAE**, we added everything and divided by the number of observations.

For **Median Absolute Error**, we **don't average them**.

Instead, sort them.

$$5,\ 5,\ 5,\ 10,\ 15$$

Now pick the middle value.

The middle value is

$$5$$

Therefore

$$\boxed{\text{MedAE}=5}$$

That's it.

---

# The formula

$$\text{MedAE}=\text{median}\left(|y_1-\hat y_1|,\ |y_2-\hat y_2|,\ ...\ |y_n-\hat y_n|\right)$$

Read it in English:

1. Compute every prediction error.
2. Ignore the sign (absolute value).
3. Sort the errors.
4. Pick the middle one.

---

# Why use the median instead of the average?

This is the important idea.

Suppose your absolute errors are

```
2, 3, 4, 5, 100
```

Imagine one prediction was terrible.

### MAE

$$\frac{2+3+4+5+100}{5}=22.8$$

The average says

> "The model is off by about 23."

But that's misleading because **four predictions were actually very good**.

---

### MedAE

Sort them

```
2,3,4,5,100
```

The middle value is

```
4
```

So

$$\boxed{\text{MedAE}=4}$$

This better represents the **typical** prediction error.

---

# Why do people say MedAE is "robust to outliers"?

Imagine you're taking an exam.

Scores:

```
80
82
85
84
0
```

One student accidentally submitted a blank paper.

### Average score

$$\frac{80+82+85+84+0}{5}=66.2$$

The average drops dramatically because of one unusual value.

### Median score

Sort them

```
0
80
82
84
85
```

Median = **82**

The median hardly changes.

The same thing happens with MedAE.

---

# Compare MAE and MedAE

Suppose your absolute errors are

```
1
2
2
3
100
```

### MAE

$$\frac{1+2+2+3+100}{5}=21.6$$

One bad prediction makes the average look much worse.

### MedAE

Sort them

```
1
2
2
3
100
```

Middle value

```
2
```

So

$$\text{MedAE}=2$$

Notice how the huge error **100** has **no effect** on the median.

---

# Intuition

Think of MAE and MedAE as asking two slightly different questions.

### MAE asks:

> **"If I average all my mistakes, how big is the average mistake?"**

Every error matters, especially very large ones.

---

### MedAE asks:

> **"What is the size of a typical mistake?"**

It doesn't let one or two terrible predictions dominate the metric.

---

## A simple way to remember all three metrics

|Metric|Question it answers|
|---|---|
|**R²**|How much of the data's variation did my model explain?|
|**MAE**|On average, how far are my predictions from the true values?|
|**MedAE**|What is the **typical** prediction error? (ignoring extreme outliers)|

A useful rule of thumb is:

- Use **MAE** when you want every error—including large ones—to contribute to the overall score.
- Use **MedAE** when your dataset may contain occasional extreme errors (outliers) and you want a measure of the model's _typical_ performance.

# [Mean Squared Error](http://scikit-learn.org/stable/modules/model_evaluation.html#mean-squared-error):

Excellent! MSE is probably **the most important error metric** in machine learning. Once you understand **why we square the errors**, MSE becomes very intuitive.

Let's explain it the same way as before.

---

# Step 1: Predict some values

Suppose you're predicting daily sales.

|Day|Actual ((y))|Predicted ((\hat y))|
|---|--:|--:|
|1|100|90|
|2|80|95|
|3|120|125|
|4|90|85|

---

# Step 2: Calculate the errors

$$
\text{Error}=y_i-\hat y_i  
$$

|Day|Error|
|---|---|
|1|10|
|2|-15|
|3|-5|
|4|5|

Just like before.

---

# Step 3: Square each error

Instead of taking the absolute value (like MAE),

we square every error.

|Error|Squared Error|
|---|---|
|10|100|
|-15|225|
|-5|25|
|5|25|

Notice something important:

- Negative errors become positive.
    
- Large errors become **much larger**.
    

---

# Step 4: Average the squared errors

Add them together.

$$
100+225+25+25=375  
$$

Now divide by the number of observations.

$$ 
MSE=\frac{375}{4}=93.75  
$$

That's your Mean Squared Error.

---

# The formula

$$
MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat y_i)^2  
$$

Read it exactly like this:

1. Compute every prediction error.
    
2. Square every error.
    
3. Add them all together.
    
4. Divide by the number of predictions.
    

In English:

> **MSE is the average of the squared prediction errors.**

---

# But why square the errors?

This is the entire reason MSE exists.

Imagine two models.

### Model A

Errors:

```text
2
2
2
2
2
```

---

### Model B

Errors:

```text
0
0
0
0
10
```

Both models have

### MAE

Model A

$$
\frac{2+2+2+2+2}{5}=2  
$$

Model B

$$
\frac{0+0+0+0+10}{5}=2  
$$

According to MAE,

**both models are equally good.**

---

But are they?

Model A is consistently a little wrong.

Model B is perfect most of the time...

...but one prediction is **terribly wrong**.

Sometimes (for example, predicting stock prices, medical diagnoses, or aircraft fuel), one huge mistake is much worse than many small mistakes.

This is where MSE shines.

---

# Let's square the errors

### Model A

Squared errors

```text
4
4
4
4
4
```

Average

$$
MSE=\frac{20}{5}=4  
$$

---

### Model B

Squared errors

```text
0
0
0
0
100
```

Average

$$
MSE=\frac{100}{5}=20  
$$

Now MSE says

> **Model B is much worse because of that one huge mistake.**

That's exactly what we wanted.

---

# Why does squaring punish large errors?

Look how quickly squaring grows.

|Error|Absolute Error|Squared Error|
|---|---|---|
|1|1|1|
|2|2|4|
|3|3|9|
|5|5|25|
|10|10|100|
|20|20|400|

Notice:

- Double the error (5 → 10)
    
- Square becomes **4 times larger** (25 → 100)
    

Large mistakes receive a much heavier penalty than small ones.

---

# Why not just use MAE?

MAE treats every error **linearly**.

Suppose two errors are

```text
2
10
```

MAE says

- Error 10 is **5× worse** than error 2.
    

MSE says

-  $$(2^2=4) $$
    
- $$ (10^2=100) $$
    

Now the large error contributes

$$
\frac{100}{4}=25  
$$

times more.

That's why people say

> **MSE penalizes large errors more heavily.**

---

# One drawback of MSE

Suppose you're predicting house prices in dollars.

Your errors are also in dollars.

But MSE squares them.

If your prediction error is

$100

then

$$
100^2=10,000  
$$

The units become

$$
\boxed{\text{dollars}^2}  
$$

which isn't very intuitive.

That's why people often compute **Root Mean Squared Error (RMSE)** afterward.

$$  
RMSE=\sqrt{MSE}  
$$

Taking the square root brings the units back to dollars, making the error easier to interpret.

---

# Compare all the metrics you've learned

|Metric|What it asks|Sensitive to outliers?|
|---|---|---|
|**MAE**|On average, how far are my predictions from the true values?|No (only linearly)|
|**MedAE**|What is the typical prediction error?|Very resistant to outliers|
|**MSE**|What is the average squared prediction error?|**Yes—large errors are heavily penalized**|
|**R²**|How much of the variation in the data did my model explain?|Not an error metric|

### An easy way to remember them

Imagine you're throwing darts at a dartboard:

- **MAE**: "On average, how many centimeters from the bullseye are my darts?"
    
- **MedAE**: "What's the distance of my _typical_ dart from the bullseye?"
    
- **MSE**: "I'm going to punish darts that land far away much more than those that are only slightly off." That's why squaring is used—it makes big misses count disproportionately more than small ones.

# [Mean Squared Logarithmic Error](http://scikit-learn.org/stable/modules/model_evaluation.html#mean-squared-logarithmic-error):

This is probably the **least intuitive** metric you've encountered so far. The sentence

> "we take the logarithm of the series"

doesn't really explain **why** we do it.

Let's build the intuition from scratch.

---

# First, remember what MSE does

MSE measures

> **the average squared error.**

$$
MSE=\frac{1}{n}\sum (y-\hat y)^2  
$$

If your errors are

```
2, 5, 10
```

MSE squares them

```
4, 25, 100
```

Large errors get punished much more than small ones.

---

# Now imagine two prediction mistakes

Suppose you're predicting company revenue.

### Prediction A

Actual:

$$
10  
$$

Predicted:

$$
20  
$$

Error = 10

---

### Prediction B

Actual:

$$
1000  
$$

Predicted:

$$
1010  
$$

Error = 10

Again, error = 10.

---

## According to MSE

Both predictions are equally bad.

Why?

Because both have

$$
(10)^2=100  
$$

---

But are they really equally bad?

Think about it.

Predicting

```
10 → 20
```

means you predicted **twice** the correct value.

That's a huge mistake.

---

Predicting

```
1000 → 1010
```

means you're only off by **1%**.

That's actually an excellent prediction.

Yet MSE gives them exactly the same penalty.

This is the main weakness of MSE.

---

# MSLE tries to fix this

Instead of comparing

```
10 vs 20
```

it compares

```
log(11) vs log(21)
```

Instead of

```
1000 vs 1010
```

it compares

```
log(1001) vs log(1011)
```

Notice the formula

## $$
MSLE=  
\frac1n  
\sum  
\left(  
\log(1+y_i)

\log(1+\hat y_i)  
\right)^2  
$$

The **+1** is simply there so we can handle values equal to zero, since $$ (\log(0)) $$ is undefined.

---

# Why logs?

This is the key idea.

A logarithm compresses large numbers.

For example

|Number|Log (approx.)|
|---|---|
|10|2.40|
|20|3.04|
|100|4.62|
|1000|6.91|
|10000|9.21|

Notice

- Numbers increased from **10 to 10,000** (a factor of 1000).
    
- Logs only increased from **2.4 to 9.2**.
    

The logarithm shrinks large values.

---

# What does that accomplish?

Instead of caring about the **absolute difference**, MSLE cares more about the **relative difference**.

Example:

### Case 1

Actual = 10

Prediction = 20

Prediction is **100% too high.**

---

### Case 2

Actual = 1000

Prediction = 1010

Prediction is only **1% too high.**

MSLE understands that these are very different situations.

MSE does not.

---

# Another example

Suppose your model predicts population.

### Prediction 1

Actual

```
50
```

Predicted

```
100
```

Error = 50

---

### Prediction 2

Actual

```
10,000
```

Predicted

```
10,050
```

Error = 50

MSE says

> Same error.

MSLE says

> No.

The first prediction doubled the value.

The second prediction missed by only 0.5%.

Those are not equally bad.

---

# Why do people say

> "MSLE gives more weight to small errors"?

That sentence is a bit misleading.

A better way to say it is:

> **MSLE gives more importance to percentage (relative) errors than absolute errors.**

For example

Going from

```
1 → 2
```

is a 100% increase.

Going from

```
1001 → 1002
```

is only a 0.1% increase.

Even though both changed by **1**, MSLE treats the first change as much more significant because it's a much larger relative change.

---

# Why is it useful for exponential data?

Imagine a company's revenue.

```
Year 1 : 100
Year 2 : 200
Year 3 : 400
Year 4 : 800
Year 5 : 1600
```

This is exponential growth.

An error of

```
50
```

means very different things over time.

At Year 1:

```
100 instead of 150
```

That's a **50%** error.

At Year 5:

```
1600 instead of 1650
```

That's only about a **3%** error.

MSLE naturally accounts for this by working on the logarithmic scale, where multiplicative (percentage) changes become additive.

---

# Compare all the metrics

|Metric|What it measures|Best for|
|---|---|---|
|**MAE**|Average absolute error|Easy interpretation|
|**MedAE**|Typical absolute error|Robust to outliers|
|**MSE**|Average squared error|Penalizing large absolute mistakes|
|**MSLE**|Average squared error on the **log scale**|Penalizing large **relative (percentage)** mistakes, especially for exponentially growing data|

## The easiest way to remember them

Imagine you're estimating salaries:

- **MSE** asks: "How many dollars was I off?"
    
    - Being off by $1,000 is treated the same whether the salary is $10,000 or $100,000.
        
- **MSLE** asks: "How many **percent** was I off?"
    
    - Being off by $1,000 on a $10,000 salary (~10%) is much worse than being off by $1,000 on a $100,000 salary (~1%).
        

That's the core intuition behind MSLE: **it measures errors on a logarithmic scale, making it much more sensitive to relative (percentage) differences than to absolute differences.**

# Mean Absolute Percentage Error:

Absolutely! MAPE is actually one of the **easiest** metrics to understand. If MAE answers

> **"On average, how many units am I wrong?"**

then MAPE answers

> **"On average, what percentage am I wrong?"**

That's why business managers love it—it speaks the language of percentages instead of raw units.

Let's build it step by step.

---

# Step 1: Remember MAE

Suppose you're predicting monthly sales.

|Month|Actual|Predicted|
|---|---|---|
|Jan|100|90|
|Feb|200|180|
|Mar|300|330|

The absolute errors are

|Error|
|---|
|10|
|20|
|30|

MAE is

$$
\frac{10+20+30}{3}=20  
$$

So we say

> **"On average, we're off by 20 sales."**

That's useful...

But is **20** a big error?

We don't know!

---

# Step 2: Why MAE can be misleading

Imagine two companies.

### Company A

Actual sales

```text
100
```

Prediction

```text
90
```

Error = 10

---

### Company B

Actual sales

```text
10,000
```

Prediction

```text
9,990
```

Error = 10

MAE says

> Both predictions are equally good.

But are they?

Not really.

For Company A,

you're off by

$$
\frac{10}{100}=10%  
$$

---

For Company B,

you're off by

$$ 
\frac{10}{10000}=0.1%  
$$

Clearly, the second prediction is much better.

This is exactly the problem MAPE solves.

---

# Step 3: Convert every error into a percentage

Instead of using

$$
|y-\hat y|  
$$

MAPE uses

$$
\frac{|y-\hat y|}{y}  
$$

In English,

> **Take the error and divide it by the actual value.**

This tells us

> **What percentage of the true value was the error?**

---

# Step 4: Example

Suppose

|Actual|Predicted|
|---|---|
|100|90|
|200|180|
|300|330|

---

### Row 1

Absolute error

$$
10  
$$

Percentage error

$$ 
\frac{10}{100}=0.10=10%  
$$

---

### Row 2

Absolute error

$$ 
20  
$$

Percentage error

$$
\frac{20}{200}=0.10=10%  
$$

---

### Row 3

Absolute error

$$
30  
$$

Percentage error

$$
\frac{30}{300}=0.10=10%  
$$

---

Average them.

$$ 
\frac{10+10+10}{3}=10%  
$$

So

$$\boxed{\text{MAPE} = 10\%}$$

Now you can say

> **"On average, our predictions are off by 10%."**

That's much easier to understand than saying

> "Our MAE is 20."

---

# The formula

$$
MAPE=  
\frac{100}{n}  
\sum  
\frac{|y_i-\hat y_i|}{y_i}  
$$

Read it exactly like this:

1. Compute the prediction error.
    
2. Ignore the sign (absolute value).
    
3. Divide by the actual value.
    
4. Convert to a percentage (multiply by 100).
    
5. Average all the percentages.
    

In plain English:

> **MAPE is the average percentage error of your predictions.**

---

# Why managers like MAPE

Imagine you're presenting your model.

### Option 1

"Our MAE is 52."

A manager might ask:

> "Is 52 good or bad?"

It depends on the scale of the data.

---

### Option 2

"Our MAPE is 4.8%."

Almost everyone immediately understands that.

> "Our forecasts are off by about 5%."

That's why MAPE is widely used in business.

---

Suppose you predict house prices.

### House A

Actual

```text
$100,000
```

Prediction

```text
$90,000
```

Absolute error

```text
$10,000
```

MAPE

$$
10%  
$$

---

### House B

Actual

```text
$1,000,000
```

Prediction

```text
$990,000
```

Absolute error

```text
$10,000
```

MAPE

$$

$$

MAE says

> Same error.

MAPE says

> The second prediction is much better because the error is a much smaller fraction of the true value.

# One important limitation

Suppose the actual value is

$$ 
y=0  
$$

Then

$$
\frac{|y-\hat y|}{y}
=
\frac{\text{something}}{0}  
$$

Division by zero is undefined.

This means **MAPE cannot be computed when actual values are zero** (and it can become extremely large when actual values are very close to zero). That's one reason other metrics, such as MAE, RMSE, or alternatives like sMAPE, are often preferred when zeros are common.

---

# Compare all the metrics you've learned

|Metric|What it asks|Units|
|---|---|---|
|**MAE**|On average, how many units am I wrong?|Same as the data|
|**MedAE**|What is the typical error?|Same as the data|
|**MSE**|Average squared error|Squared units|
|**MSLE**|Average squared error on a log scale (focuses on relative differences)|Log-squared|
|**MAPE**|On average, what percentage am I wrong?|Percentage (%)|

## The easiest way to remember all five

Imagine you're predicting monthly revenue:

- **MAE:** "I'm off by about **$500** each month."
    
- **MedAE:** "A **typical** month is off by **$450**."
    
- **MSE:** "I punish really big misses much more than small ones."
    
- **MSLE:** "I care more about **percentage differences** than dollar differences."
    
- **MAPE:** "I'm off by about **5%** on average."
    

A good mental rule is:

- **MAE** → **How many units?**
    
- **MAPE** → **How many percent?**
    
- **MSE** → **How severely should I punish large mistakes?**
    
- **MSLE** → **How wrong am I in relative (percentage) terms, especially when values span a wide range?**
    
- **MedAE** → **What does a typical mistake look like, ignoring extreme outliers?**
    

These five metrics cover most of what you'll encounter in forecasting and regression evaluation. Once you're comfortable with them, you'll be able to understand why a particular metric is chosen for a given problem.

# **SMAPE (Symmetric Mean Absolute Percentage Error)**

**SMAPE (Symmetric Mean Absolute Percentage Error)** is a metric used to measure how accurate a forecasting or prediction model is. It tells you **how far your predictions are from the actual values, expressed as a percentage**, while treating over-predictions and under-predictions more fairly than MAPE.

---

## Why do we need SMAPE?

Suppose you're predicting sales.

|Day|Actual Sales|Predicted Sales|
|---|--:|--:|
|1|100|90|
|2|200|220|
|3|150|140|

You want a single number that tells you **how good the predictions are**.

Metrics like:

- MAE → Average error in units (e.g., 13.3 sales)
    
- RMSE → Penalizes large mistakes
    
- MAPE → Average percentage error
    
- **SMAPE → Symmetric percentage error**
    

---

## The Formula

SMAPE is defined as:

$$
\boxed{  
SMAPE=\frac{100}{n}\sum_{i=1}^{n}  
\frac{|A_i-P_i|}  
{\left(\frac{|A_i|+|P_i|}{2}\right)}  
}  
$$

where:

- (A_i) = Actual value
    
- (P_i) = Predicted value
    
- (n) = Number of observations
    

An equivalent form is:

$$ 
SMAPE=  
\frac{100}{n}  
\sum  
\frac{2|A-P|}  
{|A|+|P|}  
$$

Both formulas give the same result.

---

## Step-by-Step Example

Suppose

|Actual|Predicted|
|--:|--:|
|100|90|
|200|220|
|300|330|

### Step 1: Compute the error

|Actual|Predicted|Absolute Error|
|--:|--:|--:|
|100|90|10|
|200|220|20|
|300|330|30|

---

### Step 2: Compute the average of actual and predicted

|Actual|Predicted|(Actual+Predicted)/2|
|--:|--:|--:|
|100|90|95|
|200|220|210|
|300|330|315|

---

### Step 3: Divide error by average

|Error|Average|Ratio|
|--:|--:|--:|
|10|95|0.1053|
|20|210|0.0952|
|30|315|0.0952|

---

### Step 4: Average the ratios

$$  
\frac{0.1053+0.0952+0.0952}{3}  
=0.0986  
$$

Multiply by 100:

$$
SMAPE=9.86%  
$$

**Interpretation:**

> On average, the predictions are about **9.86% away from the actual values.**

---

# Why is it called "Symmetric"?

The denominator uses **both the actual and predicted values**.

MAPE uses only:

$$
\frac{|A-P|}{A}  
$$

SMAPE uses:

$$
\frac{|A-P|}  
{(A+P)/2}  
$$

This makes the error measure **more balanced** between over-predictions and under-predictions.

---

## Example: MAPE vs SMAPE

### Case 1: Underprediction

Actual = 100

Prediction = 80

### MAPE

$$
\frac{|100-80|}{100}  
=20%  
$$

### SMAPE

$$
\frac{20}{90}  
=22.22%  
$$

---

### Case 2: Overprediction

Actual = 80

Prediction = 100

### MAPE

$$
\frac{20}{80}  
=25%  
$$

### SMAPE

$$
\frac{20}{90}  
=22.22%  
$$

Notice:

|Metric|Case 1|Case 2|
|---|--:|--:|
|MAPE|20%|25%|
|SMAPE|22.22%|22.22%|

This is why SMAPE is called **symmetric**: swapping the actual and predicted values gives the same percentage error.

---

# Another Example with Multiple Values

|Actual|Prediction|
|--:|--:|
|50|45|
|80|100|
|120|110|
|200|180|

Compute each row:

|Actual|Prediction|SMAPE (%)|
|--:|--:|--:|
|50|45|10.53|
|80|100|22.22|
|120|110|8.70|
|200|180|10.53|

Average:

$$
\frac{10.53+22.22+8.70+10.53}{4}
=
12.99%  
$$

So

**SMAPE ≈ 13%**

---

# Advantages of SMAPE

✅ Easy to interpret as a percentage.

✅ Treats overestimation and underestimation equally.

✅ Less biased than MAPE when predictions are larger than actuals.

✅ Often used in:

- Demand forecasting
    
- Sales forecasting
    
- Electricity load prediction
    
- Time-series forecasting competitions
    

---

# Limitations

### 1. Problems near zero

If both actual and predicted values are close to zero:

Actual = 0

Prediction = 0

the denominator becomes zero, making the metric undefined unless handled specially.

---

### 2. Maximum value

SMAPE ranges from

- **0%** → perfect prediction
    
- **200%** → worst possible prediction (for the common definition above)
    

For example:

Actual = 100

Prediction = 0

$$
\frac{2\times100}{100+0}  
=2  
=200%  
$$

Some libraries divide this result by 2, reporting a maximum of **100%** instead. It's important to check which convention your software uses.

---

# Comparison with Other Metrics

|Metric|Unit|Sensitive to Outliers?|Percentage?|Issue|
|---|---|---|---|---|
|MAE|Original units|No|No|Hard to compare across datasets|
|RMSE|Original units|Yes|No|Large errors dominate|
|MAPE|Percentage|No|Yes|Undefined when actual = 0|
|**SMAPE**|Percentage|No|Yes|Can still be unstable when both values are very close to zero|

---

# When should you use SMAPE?

Use SMAPE when:

- You are evaluating forecasting models.
    
- You want errors expressed as percentages.
    
- Your predictions can be both above and below the actual values.
    
- You want a metric that is less biased toward over- or under-predictions than MAPE.
    

Avoid relying on SMAPE alone if your data contains many values that are zero or extremely close to zero. In such cases, it is often helpful to report SMAPE alongside metrics like MAE or RMSE.

### Summary

- **SMAPE measures the average percentage difference between predictions and actual values.**
    
- It uses the **average of the actual and predicted values in the denominator**, making it symmetric.
    
- **Lower is better**:
    
    - **0%** = perfect predictions
        
    - Around **5–10%** = often considered very good (depending on the application)
        
    - **20%+** = may indicate substantial prediction error, though what's acceptable depends on the problem domain.


# can we overcome the limitation by adding 1 to the dinominator

Yes—you **can**, and this is actually a common practical trick. However, it changes the definition of the metric, so it is no longer the standard SMAPE.

Suppose you modify SMAPE to:

$$
\text{Modified SMAPE}
=
\frac{2|A-P|}{|A|+|P|+1}  
$$

The **+1** (sometimes called an **epsilon** or smoothing constant) ensures the denominator is never zero.

### Example 1: Both values are zero

Standard SMAPE:

- Actual = 0
    
- Prediction = 0
    

$$
\frac{2|0-0|}{0+0}
=
\frac{0}{0}  
$$

Undefined.

Modified SMAPE:

$$
\frac{2(0)}{0+0+1}=0  
$$

This makes intuitive sense: if both are zero, the error should indeed be zero.

---

### Example 2: Small values

Suppose

- Actual = 0.1
    
- Prediction = 0
    

Standard SMAPE:

$$ 
\frac{2(0.1)}{0.1}  
=2  
=200%  
$$

Modified SMAPE:

$$
\frac{0.2}{0.1+1}
=
 0.1818
=
18.18%  
$$

Notice what happened:

- Standard SMAPE says this is a **very large error**.
    
- Modified SMAPE says it's only **18%**.
    

The **+1 dominates the denominator** because the numbers themselves are tiny.

---

## Why isn't +1 part of the standard definition?

Because **1 is not scale-invariant**.

Imagine your data are measured in:

### Dollars

Actual = 1

Prediction = 0

Modified denominator:

$$
1+0+1=2  
$$

---

### Thousands of dollars

The same observation becomes

Actual = 1000

Prediction = 0

Modified denominator:

$$
1000+0+1=1001  
$$

Now the "+1" is almost irrelevant.

The metric changes simply because you changed the units, which is undesirable for a percentage-based error measure.

---

## A better approach: use a small ε

Instead of adding 1, many practitioners use a tiny constant:

$$ 
\frac{2|A-P|}  
{|A|+|P|+\varepsilon}  
$$

where

- $$ (\varepsilon = 10^{-6}) $$
    
- or  $$ (10^{-8}) $$
    
- or another very small value appropriate for the scale of the data.
    

This avoids division by zero while barely affecting the metric when values are not near zero.

---

## What do machine learning libraries do?

Different libraries handle this differently:

- **scikit-learn** does not provide SMAPE directly.
    
- Some forecasting libraries add a small epsilon internally.
    
- Others define the error for the special case ((A=P=0)) as **0** and use the standard formula otherwise.
    
- Some competitions (e.g., forecasting competitions) specify their own exact SMAPE implementation to ensure all participants are evaluated consistently.
    

---

## Recommendation

- ✅ **For research papers or benchmarking:** Use the standard SMAPE definition and clearly state how you handle the (A=P=0) case.
    
- ✅ **For production systems:** Adding a small **ε** (epsilon) is a common and reasonable engineering solution.
    
- ⚠️ **Avoid adding a fixed +1** unless your data are naturally on a scale where 1 is negligible, because it can significantly distort errors for small-valued observations.
    

In summary, **your intuition is correct**—adding something to the denominator can eliminate the division-by-zero problem. The key is that the added constant should typically be **very small relative to the data scale**, not necessarily 1, to preserve the intended behavior of SMAPE.
