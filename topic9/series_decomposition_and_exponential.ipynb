Series decomposition is one of the most important concepts in time series analysis. The idea is surprisingly simple:

> **A time series is usually made up of several different patterns mixed together. Decomposition separates those patterns.**

Think of a song. When you listen to it, you hear one sound, but it's actually made of:

* vocals
* drums
* guitar
* piano

Series decomposition does the same thing for data.

---

## Example

Suppose a coffee shop records monthly sales.

| Month | Sales ($) |
| ----- | --------: |
| Jan   |    10,000 |
| Feb   |    10,500 |
| Mar   |    11,000 |
| Apr   |    12,500 |
| May   |    13,000 |
| Jun   |    14,000 |
| Jul   |    16,000 |
| Aug   |    15,500 |
| Sep   |    14,500 |
| Oct   |    13,000 |
| Nov   |    12,000 |
| Dec   |    18,000 |

Looking at the numbers, several things are happening simultaneously.

---

# 1. Trend

The overall direction.

```
10000
10500
11000
12500
13000
14000
16000
...
```

Ignoring the small ups and downs,

```
      /
     /
    /
   /
  /
```

Sales are generally increasing.

This is called the **trend**.

---

# 2. Seasonality

Some events repeat regularly.

For example:

* Every December sales increase.
* Every July tourism increases.
* Every Monday traffic is heavier.

```
Jan  ↓
Feb
Mar
Apr
May
Jun
Jul ↑
Aug
Sep
Oct
Nov
Dec ↑↑↑
```

These repeating patterns are called **seasonality**.

---

# 3. Noise (Random variation)

Some changes happen randomly.

For example,

A rainy day.

A holiday.

A celebrity mentions your business.

These cannot be predicted perfectly.

```
100
101
99
103
97
102
```

These random fluctuations are called **noise** or **residuals**.

---

# 4. The decomposition idea

Instead of seeing one complicated graph,

```
Sales
 ^
 |
 |      /\__
 |    _/    \__
 | __/
 +------------------>
```

we split it into

```
Original
      │
      ▼
--------------------

Trend
      │
      ▼
↗↗↗↗↗↗

Seasonality
      │
      ▼
↑ ↓ ↑ ↓ ↑ ↓

Residual
      │
      ▼
small random bumps
```

Now each component is much easier to understand.

---

# Additive decomposition

Sometimes the parts simply add together.

[
\text{Observed} = \text{Trend} + \text{Seasonality} + \text{Residual}
]

Example:

```
Trend       = 100
Seasonality = +20
Noise       = -5

Observed = 100 + 20 - 5 = 115
```

This works well when seasonal effects stay about the same size over time.

Example:

```
Trend

100
110
120
130

Seasonality

+10
-10
+10
-10
```

The seasonal swing is always about ±10.

---

# Multiplicative decomposition

Sometimes the seasonal effect grows as the series grows.

Then

[
\text{Observed} =
\text{Trend}
\times
\text{Seasonality}
\times
\text{Residual}
]

Example:

```
Trend = 100

December multiplier = 1.5

Observed = 150
```

A few years later

```
Trend = 1000

December multiplier = 1.5

Observed = 1500
```

The holiday boost becomes larger because it's proportional to the current level.

---

# Why do we decompose?

Suppose you want to forecast next month's sales.

Without decomposition, the data looks messy.

```
120
150
130
170
160
210
190
```

You don't know what's causing the changes.

After decomposition:

```
Trend:
100
110
120
130
140

Seasonality:
+20
-10
+20
-10
+20

Residual:
small random errors
```

Now you can:

* forecast the trend,
* add the expected seasonal effect,
* ignore the random noise.

This often produces better forecasts and helps you understand *why* the data behaves as it does.

---

# How does this relate to smoothing?

These concepts are closely connected:

* **Smoothing** reduces random fluctuations to reveal the underlying level or trend.
* **Decomposition** goes further by separating the series into **trend**, **seasonality**, and **residual** components.

You can think of it like this:

```
Raw Time Series
       │
       ▼
Smoothing
(Removes some noise)
       │
       ▼
Decomposition
       │
       ├── Trend
       ├── Seasonality
       └── Residual
```

---

## An analogy

Imagine you're watching the sea.

The water level changes because of three things:

* 🌊 **Tide** → a slow, long-term rise and fall (**trend**)
* 🌬️ **Waves** → regular repeating motion (**seasonality**, in the analogy)
* 💨 **Wind splashes** → irregular disturbances (**noise**)

From the shore, you only see the combined water level. Decomposition is like having instruments that tell you:

* how much is due to the tide,
* how much is due to the waves,
* how much is just random splashing.

Once you've separated these effects, understanding and predicting the water level becomes much easier.


# Double Exponential Smoothing

Absolutely. Double Exponential Smoothing (also called **Holt's Linear Trend Method**) sounds complicated because of the equations, but the idea is very simple.

---

# First, why do we need it?

Simple Exponential Smoothing assumes the data has **no trend**.

For example:

```text
10
11
10
11
10
11
```

The values fluctuate around 10–11.

Predicting the next value as "about 10.5" makes sense.

---

But what if the data keeps increasing?

```text
10
12
14
16
18
20
```

If Simple Exponential Smoothing predicts

```text
Next = 18
```

that's not very good.

The data is clearly going **up**.

We need to learn **how fast** it's increasing.

---

# The main idea

Instead of learning only

> "Where is the series now?"

we also learn

> "Which direction is it moving?"

So we keep track of **two things**:

### 1. Level (ℓ)

Think of it as

> **Current value of the series**

Example:

```text
Sales today ≈ 100
```

Level = 100

---

### 2. Trend (b)

Think of it as

> **How much the value changes every step**

Example

```text
100
105
110
115
120
```

The trend is about

```text
+5 per day
```

So

```text
Level = 120

Trend = +5
```

---

# Then the prediction becomes easy

Instead of saying

```text
Tomorrow = 120
```

we say

```text
Tomorrow = 120 + 5

=125
```

That's exactly what the last equation says.

[
\hat y_{x+1}=l_x+b_x
]

which simply means

> **Prediction = Current level + Current trend**

---

# Understanding each equation

---

## Equation 1

[
l_x=\alpha y_x+(1-\alpha)(l_{x-1}+b_{x-1})
]

Ignore the symbols.

Read it like this:

> **New level = Some of today's actual value + Some of yesterday's prediction**

Suppose yesterday

```text
Level = 100

Trend = +5
```

So yesterday predicted today would be

```text
100 + 5 =105
```

But today's actual value is

```text
108
```

Should we completely trust today's value?

No.

Should we completely trust yesterday's prediction?

No.

So we mix them.

Example

```text
alpha =0.4
```

New level

```text
=0.4×108

+0.6×105

=43.2+63

=106.2
```

So our new estimate of today's level becomes

```text
106.2
```

---

## Equation 2

[
b_x=\beta(l_x-l_{x-1})+(1-\beta)b_{x-1}
]

This updates the trend.

The first part

```text
l_x-l_{x-1}
```

means

> **How much did the level change?**

Example

Yesterday's level

```text
100
```

Today's level

```text
106
```

Difference

```text
6
```

So maybe the trend is +6.

But yesterday we thought the trend was

```text
+5
```

Again, don't trust only one observation.

Mix both.

Suppose

```text
beta=0.3
```

New trend

```text
=0.3×6

+0.7×5

=1.8+3.5

=5.3
```

So instead of suddenly jumping from 5 to 6,

the trend slowly changes to

```text
5.3
```

---

## Equation 3

Now predict tomorrow

Current level

```text
106.2
```

Current trend

```text
5.3
```

Prediction

```text
106.2+5.3

=111.5
```

Very simple.

---

# Think of driving a car

Imagine you're driving.

The **level** is

```text
Where is the car now?
```

The **trend** is

```text
How fast is it moving?
```

If you're at

```text
100 km
```

and driving

```text
60 km/h
```

then after one hour you expect

```text
160 km
```

Exactly the same idea.

---

# Why do we smooth the trend too?

Suppose your observed increases are

```text
+5
+4
+7
+6
+5
```

The trend itself is noisy.

Instead of believing every increase exactly,

Double Exponential Smoothing smooths these changes too.

That's why we have another parameter,

```text
β (beta)
```

* **α** smooths the **level**
* **β** smooths the **trend**

---

# Summary

| Component      | Meaning                          | Example |
| -------------- | -------------------------------- | ------- |
| (y_x)          | Actual observation               | 108     |
| (\ell_x)       | Smoothed current level           | 106.2   |
| (b_x)          | Smoothed trend (change per step) | +5.3    |
| (\hat y_{x+1}) | Prediction                       | 111.5   |

So, compared to **Simple Exponential Smoothing**, which only asks:

> "What's the current value?"

**Double Exponential Smoothing** asks two questions:

1. **What's the current value?** (level)
2. **How fast is it changing?** (trend)

By keeping track of both, it can forecast data that is steadily increasing or decreasing much more accurately.


This paragraph is basically explaining what the two parameters (**α** and **β**) control. Here's a simpler version.

---

## In Double Exponential Smoothing, there are two things to estimate:

1. **Level (current value)** → controlled by **α (alpha)**
2. **Trend (rate of increase/decrease)** → controlled by **β (beta)**

---

## What does α do?

Recall the level update:

[
l_x=\alpha y_x+(1-\alpha)(l_{x-1}+b_{x-1})
]

It decides **how much you trust the newest observation** when estimating the current level.

### Large α (e.g., 0.9)

You trust the latest data a lot.

Example:

```
Yesterday's estimated level = 100
Today's actual value = 120
```

With α = 0.9:

```
New level ≈ 0.9×120 + 0.1×100 = 118
```

The estimate jumps quickly toward the new value.

---

### Small α (e.g., 0.1)

You trust the past more.

```
New level ≈ 0.1×120 + 0.9×100 = 102
```

The estimate changes slowly.

So:

* **Large α → responds quickly**
* **Small α → smoother level**

---

## What does β do?

β controls **how quickly the trend changes**.

Suppose you thought the trend was

```
+5 units/day
```

but the new data suggests it should be

```
+8 units/day
```

### Large β

You update the trend quickly.

```
Old trend = 5

New trend ≈ 7.7
```

---

### Small β

You update the trend cautiously.

```
Old trend = 5

New trend ≈ 5.3
```

So:

* **Large β → trend changes quickly**
* **Small β → trend changes slowly**

---

## Why does the tutorial say:

> "The larger the values, the more weight the most recent observations will have."

Because:

* **α** determines how much the **latest observation** affects the **level**.
* **β** determines how much the **latest change in level** affects the **trend**.

Higher values mean "listen more to recent data."

Lower values mean "trust the historical pattern."

---

## Why can some parameter combinations give strange results?

Imagine noisy data:

```
100
105
99
107
101
106
```

If you choose:

```
α = 0.99
β = 0.99
```

Then every tiny fluctuation changes both the level and the trend dramatically.

The model might think:

```
Trend = +8
Next point...
Trend = -5
Next point...
Trend = +10
```

The forecast becomes unstable because it's chasing every random bump.

On the other hand, if you choose:

```
α = 0.01
β = 0.01
```

the model barely reacts to genuine changes. If the series starts increasing steadily, the estimated level and trend will lag far behind.

---

## So how do we choose α and β?

Instead of guessing, we let the computer try many combinations (for example, α = 0.1, β = 0.2; α = 0.3, β = 0.8; etc.) and measure which pair produces the smallest prediction error on historical data. The pair with the lowest error is selected.

That's what the tutorial means by:

> "We'll look into choosing parameters automatically."

---

### A simple analogy

Imagine you're driving a car:

* **α** is how quickly you adjust the **steering wheel** based on what you currently see.
* **β** is how quickly you change your **speed** when you notice the road is going uphill or downhill.

If both are **too high**, you're constantly overcorrecting.

If both are **too low**, you're slow to react.

A good balance lets the model follow the true path smoothly without chasing every small fluctuation.
