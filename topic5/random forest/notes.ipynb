## What does random subspace mean?

### What is the random subspace method?

The random subspace method means:

> **Instead of allowing a decision tree to consider all features (columns) when deciding a split, it is allowed to consider only a random subset of the features.**

For example, suppose your dataset has **10 features**:

```text
Age
Salary
Height
Weight
Gender
Education
Experience
City
Marital Status
Income
```

Normally, a decision tree considers **all 10 features** at every split.

With the **random subspace method**, at each split it randomly picks, say, **3 features**:

```
Split 1:
Randomly selected features:
✓ Salary
✓ Height
✓ Experience

The tree chooses the best split only from these 3.
```

For the next split, it randomly selects another subset:

```
Split 2:
✓ Age
✓ Income
✓ Education

Again, it chooses the best split only from these.
```

Notice that the subset changes **at every split**.

---

### Why do this?

Imagine one feature is extremely informative.

Without the random subspace method:

```
Tree 1 → first split = Income
Tree 2 → first split = Income
Tree 3 → first split = Income
Tree 4 → first split = Income
```

Even though each tree was trained on a different bootstrap sample, they all look very similar.

Highly similar trees are **highly correlated**.

Bagging works best when the trees make different mistakes.

---

With the random subspace method:

```
Tree 1:
First split can only choose from
{Age, Height, Weight}

Tree 2:
First split can only choose from
{Income, Experience, Gender}

Tree 3:
First split can only choose from
{Salary, Education, City}
```

Now every tree grows differently.

This reduces the **correlation** between the trees.

---

### Why does reducing correlation help?

Suppose you have 100 identical trees.

```
Tree 1: Cat
Tree 2: Cat
...
Tree100: Cat
```

Voting doesn't improve anything because they're all making the same prediction.

Now suppose the trees are different:

```
Tree 1: Cat
Tree 2: Dog
Tree 3: Cat
Tree 4: Cat
Tree 5: Dog
...
```

Their errors are less likely to occur on the same examples. When you combine their predictions using majority voting, many individual mistakes cancel out, leading to a better overall model.

---

### Bagging vs. Random Forest

This is the distinction your sentence is referring to:

| Bagging                                                     | Random Forest                                                          |
| ----------------------------------------------------------- | ---------------------------------------------------------------------- |
| Randomly samples **training examples** (bootstrap sampling) | Randomly samples **training examples** **and** **features**            |
| Trees can use all features at each split                    | Trees use only a random subset of features at each split               |
| Reduces variance                                            | Further reduces correlation between trees, often improving performance |

So, the **random subspace method** is the extra idea that turns **bagging with decision trees** into a **Random Forest**. It forces each tree to look at different subsets of features, making the trees more diverse and improving the effectiveness of their combined predictions.

# what is n_min

Let's ignore decision trees for a moment and think about a simple analogy.

### Imagine you're dividing a class of students

You have **20 students**.

You keep dividing them into smaller groups.

Question: **When should you stop dividing?**

You decide on a rule:

> **"Never make a group with fewer than 4 students."**

Here, **4 is (n_{\text{min}})**.

So:

```text
20 students
├── 10
│   ├── 5
│   │   ├── 3 ❌ Not allowed (less than 4)
│   │   └── 2 ❌ Not allowed
│   └── 5
└── 10
```

Since splitting a group of 5 would create groups smaller than 4, you **stop**.

---

## Now apply this to a decision tree

Suppose a node contains **12 training samples**.

If you set:

[
n_{\text{min}} = 5
]

then:

```
Node: 12 samples
```

The algorithm asks:

> "Can I split these 12 samples into two child nodes **without violating the minimum size rule**?"

Possible split:

```
12
├── 7
└── 5
```

✅ Allowed because both children have at least 5 samples.

Another possible split:

```
12
├── 10
└── 2
```

❌ Not allowed because one child has only 2 samples (< 5).

---

## Another example

Suppose a node has **6 samples**, and

[
n_{\text{min}} = 5
]

Can it split?

Possible splits:

```
6 → 3 + 3
```

No (both are below 5).

```
6 → 5 + 1
```

No (1 is below 5).

So **the node becomes a leaf**.

---

## Why do we need this?

If we don't have this rule, the tree could keep splitting until every leaf contains just **one sample**:

```
100
├── 50
│   ├── 25
│   │   ├── ...
│   │   │   ├── 1
│   │   │   └── 1
```

This usually memorizes the training data (**overfitting**).

Setting (n_{\text{min}}) prevents the tree from creating tiny leaves.

---

### One-sentence definition

> **(n_{\text{min}}) is the minimum number of training samples that must be present in a leaf (or child node, depending on the algorithm) after a split. If a split would create a smaller node than this, the algorithm does not perform the split.**

---

**One important note:** Different books define (n_{\text{min}}) slightly differently. Some mean:

* **minimum samples in a leaf** (`min_samples_leaf`),

while others mean:

* **minimum samples required before attempting to split a node** (`min_samples_split`).
