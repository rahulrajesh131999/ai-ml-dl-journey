This section is basically explaining:

"Why do we minimize Mean Squared Error in linear regression?"

And the answer is:

Because we assume the errors are Normally distributed.

That's the entire story.

Let's unpack it slowly because this is where:

* statistics,
* probability,
* and ML

all connect together.

# 🎯 Big Picture

Earlier, linear regression looked like:

$$y = Xw + \epsilon$$

which meant:

$$\text{actual value} = \text{prediction} + \text{noise}$$

Now the article asks:

"What if we model that noise probabilistically?"

That leads us to Maximum Likelihood Estimation (MLE).

# 🧠 Step 1 — The Model

The linear model is:

$$y_i = w^Tx_i + \epsilon_i$$

where:

* $x_i$ = feature vector
* $w$ = weights
* $w^Tx_i$ = prediction
* $\epsilon_i$ = random error

# 🔥 Important New Assumption

Now they assume:

$$\epsilon_i \sim \mathcal{N}(0,\sigma^2)$$

Meaning:
The random variable ei is distributed normally with mean 0 and variance sigma2

Errors follow a Normal (Gaussian) distribution.

# 🧠 What does that mean intuitively?

Most prediction errors are:

* small,
* centered around zero,
* with occasional larger mistakes.

Like this:

$$y = e^{-x^2}$$

Bell-shaped curve.

# 🎯 Key Insight

Since:

$$y_i = w^Tx_i + \epsilon_i$$

and:

$$\epsilon_i \sim \mathcal{N}(0,\sigma^2)$$

then:

$$y_i \sim \mathcal{N}(w^Tx_i,\sigma^2)$$

This is VERY important.

It means:

Each target value is normally distributed around the model prediction.

# 🧠 Intuition

Suppose the model predicts:

$$\hat y = 50$$

Reality might be:

* 49
* 51
* 52
* 48

because of noise.

So:

$$y_i \sim \mathcal{N}(50,\sigma^2)$$

# 🎯 Step 2 — Probability of Observing Data

Now we ask:

"How probable is our observed data under these weights?"

This is called the likelihood.

For one point:

$$p(y_i|X;w)$$

means:

Probability of observing $y_i$ given:

* features $X$
* weights $w$

Since:

$$y_i \sim \mathcal{N}(w^Tx_i,\sigma^2)$$

we use Gaussian density:

$$p(y_i|X;w) = \mathcal{N}(w^Tx_i,\sigma^2)$$

# 🧠 Step 3 — Full Dataset Likelihood

We assume samples are independent.

So total probability becomes:

$$\prod_{i=1}^n p(y_i|X;w)$$

Meaning:

Multiply probabilities of all observations.

# ❓Why use log-likelihood?

Because products are ugly mathematically.

Example:

$$0.001 \times 0.0003 \times 0.002$$

Hard to work with.

Logs convert products → sums:

$$\log(ab)=\log a+\log b$$

Much easier.

# 🎯 Step 4 — The Gaussian Formula

Gaussian density is:

$$\mathcal{N}(\mu,\sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(y-\mu)^2}{2\sigma^2}}$$

Here:

* mean = $w^Tx_i$
* variance = $\sigma^2$

After taking log, the ugly exponential becomes:

$$-\frac{1}{2\sigma^2} (y_i-w^Tx_i)^2$$

Summing all points gives:

$$-\frac{1}{2\sigma^2} \sum (y_i-w^Tx_i)^2$$

# 🔥 THIS IS THE HUGE MOMENT

Look carefully:

$$\sum (y_i-w^Tx_i)^2$$

That is exactly:

# 🎯 Mean Squared Error / Least Squares

# 🧠 What does this mean?

Maximizing likelihood becomes equivalent to:

$$\min \sum (y_i-w^Tx_i)^2$$

So:

# 🔥 Linear Regression + Gaussian Noise ⇒ MSE Loss

# 🎯 Why can they remove constants?

The article removes terms like:

$$-\frac{n}{2} \log 2\pi\sigma^2$$

because they do NOT depend on $w$.

When optimizing, constants don't matter.

Example:

Maximize:

$$100 - x^2$$

is same as maximizing:

$$-x^2$$

because 100 never changes.

# 🧠 Final Conclusion

The article proves:

$$\text{MLE} \equiv \text{Least Squares}$$

under Gaussian noise assumptions.

# 🔥 The Deep Intuition

This is the REAL reason MSE exists.

Not because:

"someone invented it randomly"

But because:

If errors are Normally distributed, then maximizing data probability naturally leads to squared error minimization.

# 🧠 Why Squared Error Specifically?

Because Gaussian distributions contain:

$$e^{-x^2}$$

The square comes directly from the Gaussian equation.

# 🎯 SUPER IMPORTANT CONNECTION

Different probability assumptions lead to different loss functions.

Gaussian Noise

$$\epsilon \sim \mathcal{N}$$

gives:

$$\text{MSE}$$

Laplace Noise

Would give:

$$|y-\hat y|$$

(MAE loss)

# 🧠 Final Mental Model

Linear regression is secretly saying:

"I believe the world is linear, and the remaining errors are random Gaussian noise."

Then MLE says:

"Find weights that make the observed data most probable."

That mathematically becomes:

$$\min \sum (y_i-\hat y_i)^2$$

which is Ordinary Least Squares.

Machine Learning losses are NOT arbitrary.
