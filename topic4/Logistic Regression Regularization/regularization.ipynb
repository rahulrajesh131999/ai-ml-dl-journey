This is the objective function used in many machine learning models (such as regularized logistic regression and SVMs):
$$J(X,y,w)=\mathcal{L}+\frac{1}{C}|w|^2$$
Let's break it apart.
### 1. The loss term $\mathcal{L}$
This measures **how badly the model fits the training data**.
* Small loss = predictions are close to the true labels.
* Large loss = many mistakes or poor predictions.

The model wants to make this term as small as possible.

---
### 2. The regularization term $|w|^2$
$$|w|^2 = w_1^2+w_2^2+\cdots+w_n^2$$
This measures the size of the weights.
* Large weights ⇒ large penalty.
* Small weights ⇒ small penalty.

Regularization encourages the model to keep weights small and avoid overly complex decision boundaries.

---
### 3. What does $C$ do?
The coefficient in front of the regularization term is
$$\frac{1}{C}.$$
* **Large $C$** ⇒ $\frac{1}{C}$ is small ⇒ weak regularization.
* **Small $C$** ⇒ $\frac{1}{C}$ is large ⇒ strong regularization.

So:

| C          | Regularization |
| ---------- | -------------- |
| Very large | Weak           |
| Medium     | Moderate       |
| Very small | Strong         |

---
### The tug-of-war
The model is trying to balance:
* Fit the training data well ($\mathcal{L}$ small)
* Keep weights small ($|w|^2$ small)

Think of it as:
> "I want to explain the data, but I don't want an unnecessarily complicated model."

---
## Does increasing regularization increase training accuracy?
Usually **no**.

In fact, stronger regularization often **decreases training accuracy**.

Why?

Because you're restricting the model.

Imagine a flexible curve that can perfectly fit all training points. Strong regularization says:
> "Don't bend so much. Keep the weights small."

As a result, the model may not fit the training data as closely.

Typical behavior:

| Regularization Strength | Training Accuracy |
| ----------------------- | ----------------- |
| Very weak               | High              |
| Moderate                | Slightly lower    |
| Very strong             | Much lower        |

---
## Then why use regularization?
Because we care about **test accuracy**, not just training accuracy.

Without regularization:
* Training accuracy may be 100%.
* Test accuracy may be poor (overfitting).

With some regularization:
* Training accuracy may drop a little.
* Test accuracy often improves.

Example:

| Model                   | Train Acc | Test Acc |
| ----------------------- | --------- | -------- |
| No regularization       | 99%       | 75%      |
| Moderate regularization | 95%       | 90%      |
| Too much regularization | 70%       | 68%      |

So regularization often **hurts training accuracy but helps generalization**.

---
A useful rule to remember:
> **Increasing $C$** → weaker regularization → training accuracy tends to increase.
>
> **Decreasing $C$** → stronger regularization → training accuracy tends to decrease.

That's why when tuning models, people often search over different values of $C$ to find the best balance between underfitting and overfitting.

# How Regularization effects the accuracy score

What you're seeing is **exactly what we would expect**.
Remember:
$$J = \mathcal{L} + \frac{1}{C}|w|^2$$
As $C$ increases:
$$\frac{1}{C}$$
gets smaller, so the regularization penalty becomes weaker.

Let's look at your values:

| C       | Regularization strength |
| ------- | ----------------------- |
| $0.01$  | Very strong             |
| $1$     | Moderate                |
| $10000$ | Almost none             |
---
### Case 1: $C = 0.01$
Strong regularization.

The model is heavily punished for having large weights.

It therefore chooses a very simple decision boundary.

Result:
* Simpler model
* Many training points misclassified
* Training accuracy = **62.7%**
---
### Case 2: $C = 1$
Less regularization.

The model is allowed to use larger weights and a more flexible boundary.

Result:
* Better fit to training data
* Accuracy = **83.1%**
---
### Case 3: $C = 10000$
Almost no regularization.

The model can freely adjust weights to fit the training data.

Result:
* Very flexible boundary
* Accuracy = **87.3%**
---
So when you ask:
> why did the accuracy increase when C increases?

Because **increasing $C$ means decreasing regularization**.

And decreasing regularization usually allows the model to fit the training set more closely.

In fact, if we sketch the trend:

```text
Strong reg.                        Weak reg.
      |                                 |
C=0.01 ---------> C=1 ---------> C=10000
62.7%              83.1%           87.3%
```
---
### The subtle point
Earlier you asked:
> does increasing regularization increase training accuracy?

The answer was **usually no**.

Now notice:
* Increasing regularization = decreasing $C$
* Decreasing regularization = increasing $C$

Your experiment increased **$C$**, not regularization.

So your results are perfectly consistent with the theory.
---
### Why doesn't everyone just use $C=10000$?
Because training accuracy is not the goal.

A model with almost no regularization may memorize peculiarities of the training data and perform worse on unseen data.

For example:

| C     | Train Accuracy | Test Accuracy |
| ----- | -------------- | ------------- |
| 0.01  | 63%            | 65%           |
| 1     | 83%            | 88%           |
| 10000 | 87%            | 80%           |

The training accuracy keeps going up, but the test accuracy may eventually go down due to **overfitting**.

That's why $C$ is usually chosen using a validation set or cross-validation, not by maximizing training accuracy.
