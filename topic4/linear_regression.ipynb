# 🎯 THE BIG IDEA

Linear regression tries to answer:

> "Can we mathematically find the best linear relationship between inputs and outputs?"

Example:

| Size | Price |
| ---- | ----- |
| 1000 | 2L    |
| 1200 | 2.5L  |
| 1500 | 3L    |

We want a formula like:

$$\text{price} = w_0 + w_1(\text{size})$$

That's all linear regression is.

# 🧠 PART 1 — The Linear Model

The article starts with:

$$y = w_0 + \sum_{i=1}^m w_i x_i$$

This means:

$$y = w_0 + w_1x_1 + w_2x_2 + ... + w_mx_m$$

## 🔹 Meaning

| Symbol | Meaning    |
| ------ | ---------- |
| $x_i$  | features   |
| $w_i$  | weights    |
| $y$    | prediction |

# 🎯 Example

Suppose:

| Feature | Meaning    |
| ------- | ---------- |
| $x_1$   | house size |
| $x_2$   | bedrooms   |

Then:

$$y = w_0 + w_1x_1 + w_2x_2$$

might become:

$$\text{price} = 50000 + 200(\text{size}) + 10000(\text{bedrooms})$$

# 🧠 PART 2 — Why Add $x_0 = 1$?

The article says:

$$x_0 = 1$$

This is just a mathematical trick.

Instead of:

$$w_0 + w_1x_1 + w_2x_2$$

we write:

$$w_0x_0 + w_1x_1 + w_2x_2$$

because:

$$x_0 = 1$$

Now everything becomes one clean summation.

# 🔥 Dot Product Form

Then they write:

$$y = w^Tx$$

This is just compact notation.

It means:

$$w^Tx = w_0x_0 + w_1x_1 + ... + w_mx_m$$

# 🧠 PART 3 — Matrix Form

Instead of writing equations one-by-one, we combine ALL data into matrices.

This becomes:

$$y = Xw + \epsilon$$

This is THE core linear regression equation.

# 🔹 Meaning of each term

## 1. $X$ → Feature Matrix

Contains all training examples.

Example:

$$X =
\begin{bmatrix}
1 & 1000 & 2 \\
1 & 1200 & 3 \\
1 & 1500 & 4
\end{bmatrix}$$

Rows = observations  
Columns = features

First column = ones (bias term)

## 2. $w$ → Weight Vector

$$w =
\begin{bmatrix}
w_0 \\
w_1 \\
w_2
\end{bmatrix}$$

These are what we want to learn.

## 3. $y$ → Target Vector

$$y =
\begin{bmatrix}
200000 \\
250000 \\
300000
\end{bmatrix}$$

Actual outputs.

## 4. $\epsilon$ → Noise/Error

Reality is messy.

Maybe:

* market fluctuated,
* measurement error,
* hidden factors.

So:

$$y = Xw + \epsilon$$

means:

> actual value = prediction + noise

# 🧠 PART 4 — Assumptions About Errors

These are the Gauss-Markov assumptions.

# 🔹 1. Mean Zero

$$E[\epsilon_i]=0$$

Meaning:

Errors average out to zero.

Good:

* sometimes prediction too high
* sometimes too low

Bad:

* always too high

# 🔹 2. Equal Variance (Homoscedasticity)

$$Var(\epsilon_i)=\sigma^2$$

Meaning:

Noise level should stay similar everywhere.

---

Good:

```
small random spread everywhere
```

Bad:

```
small spread initially, huge spread later
```

# 🔹 3. Errors Uncorrelated

$$Cov(\epsilon_i,\epsilon_j)=0$$

One error shouldn't influence another.

# 🧠 PART 5 — Linear Estimator

The article says:

$$\hat w_i = \omega_{1i}y_1 + \omega_{2i}y_2 + ...$$

This means:

> Estimated weights are linear combinations of outputs.

That's why it's called:

# 🔥 Linear Regression

# 🧠 PART 6 — Unbiased Estimator

They define:

$$E[\hat w_i]=w_i$$

Meaning:

> On average, the estimated weight equals the true weight.

# 🎯 Intuition

Suppose true weight is:

$$w=5$$

Your estimates across many datasets:

$$4.9,\ 5.1,\ 5.0,\ 5.2,\ 4.8$$

Average ≈ 5

So estimator is unbiased.

# 🧠 PART 7 — The Loss Function

Now comes the MOST important part.

They define:

$$\mathcal{L} = \frac{1}{2n} \sum (y_i-w^Tx_i)^2$$

This is:

# 🔥 Mean Squared Error (MSE)

# 🔹 What does it mean?

Prediction:

$$\hat y_i=w^Tx_i$$

Error:

$$y_i-\hat y_i$$

Square it:

$$(y_i-\hat y_i)^2$$

Add all errors.

# ❓Why square?

Because:

$$-5 + 5 = 0$$

Errors cancel otherwise.

Squaring:

* removes negatives
* punishes large errors heavily

# 🧠 Matrix Form of Loss

They rewrite:

$$|y-Xw|^2$$

This is just compact matrix notation for:

$$\sum (y_i-\hat y_i)^2$$

# 🧠 PART 8 — Optimization

Goal:

> Find weights minimizing error.

So we:

1. Differentiate loss wrt $w$
2. Set derivative = 0
3. Solve

Exactly like normal calculus.

# 🔥 Critical Derivation

They get:

$$\frac{\partial \mathcal{L}}{\partial w} = 0$$

which simplifies to:

$$X^TXw=X^Ty$$

called the:

# 🎯 Normal Equation

# 🧠 Final Solution

Solving for $w$:

$$w=(X^TX)^{-1}X^Ty$$

This is THE closed-form solution for linear regression.

# 🔥 Meaning of Each Piece

## $X^T$

Transpose matrix.

Rows ↔ columns.

## $X^TX$

Captures relationships between features.

## $(X^TX)^{-1}$

Matrix inverse.

Acts like division.

## Final Result

Produces optimal weights minimizing MSE.

# 🧠 PART 9 — Why "Full Column Rank"?

The article says:

$$rank(X)=m+1$$

Meaning:

Columns must be independent.

Otherwise:

* inverse doesn't exist
* solution becomes unstable

# 🧠 PART 10 — Gauss-Markov Theorem

This theorem says:

Under those assumptions:

$$w=(X^TX)^{-1}X^Ty$$

is the:

# 🔥 BEST Linear Unbiased Estimator (BLUE)

Meaning:

* linear
* unbiased
* lowest variance

among all linear unbiased estimators.

# 🎯 The Entire Story in Plain English

Linear regression says:

## Step 1

Assume output is approximately linear.

## Step 2

Measure prediction error.

## Step 3

Square errors.

## Step 4

Find weights minimizing total squared error.

## Step 5

Use calculus → derive:

$$w=(X^TX)^{-1}X^Ty$$

# 🧠 MOST IMPORTANT THINGS TO UNDERSTAND

Don't memorize matrix derivatives yet.

Focus on understanding:

* what $X$, $w$, $y$ represent
* why errors are squared
* why optimization is needed
* what the normal equation does intuitively

That's the real foundation.

# 🔥 One Important Modern ML Insight

In real-world ML:

* datasets are huge
* matrix inverse is expensive

So modern systems often use:

* Gradient Descent
* Stochastic Gradient Descent

instead of directly computing:

$$(X^TX)^{-1}$$

But mathematically, OLS is still foundational.
