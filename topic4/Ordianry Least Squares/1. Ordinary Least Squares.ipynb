## We will start studying linear models with linear regression. First of all, you must specify a model that relates the dependent variable y to the explanatory factors (or features); 

Think of linear regression as trying to draw the "best straight-line relationship" between some inputs and an output.

Suppose you want to predict a student's exam score.

* The exam score is the dependent variable (y).
* Things that may affect the score — like hours studied, attendance, sleep, etc. — are the features (or explanatory variables).

The model says:
"The output (y) is made from:

* a base value,
* plus contributions from each feature,
* plus some randomness we cannot explain."

For one feature, the model is:

$$y = \beta_0 + \beta_1 x + \varepsilon$$

Here is the meaning in plain language:

* $y$ → what we want to predict (example: exam score)
* $x$ → an input feature (example: hours studied)
* $\beta_0$ → the starting value (score expected even if $x=0$)
* $\beta_1$ → how strongly $x$ affects $y$ (how much the score increases per extra study hour)
* $\varepsilon$ → everything we cannot perfectly explain (luck, stress, guessing, illness, etc.)

Example:

Suppose the fitted model becomes:

$$\text{Score} = 40 + 5(\text{Hours Studied})$$

This means:

* A student who studies 0 hours is predicted to score 40.
* Every extra hour studied increases the predicted score by 5 points.

So:

* 2 hours studied → predicted score: $40 + 5(2) = 50$
* 6 hours studied → predicted score: $40 + 5(6) = 70$

Now imagine several features:

* hours studied
* attendance
* sleep
* assignments completed

Then the model becomes:

$$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_p x_p + \varepsilon$$

Meaning:

* each feature contributes something to the final prediction,
* and each coefficient $\beta$ tells us how important that feature is.

The main goal of linear regression is:

Find the values of the coefficients ($\beta$'s) that make predictions as close as possible to the real data.

So the algorithm looks at many data points and finds the line (or hyperplane) that best fits them.

> # Introduction

# Ordinary Least Squares (OLS) — A Visual Walkthrough

> *How do we mathematically find the best straight-line relationship between inputs (features) and outputs (targets)?*

---

## 1. The Core Idea of Linear Regression

Suppose you want to predict house prices. Your features might be: size, number of rooms, age of house.

A linear model assumes the prediction is a **weighted sum of features**:

$$\hat{y} = w_0 + w_1x_1 + w_2x_2 + \dots + w_px_p$$

Where:
- $x_i$ = feature values
- $w_i$ = coefficients/weights
- $\hat{y}$ = predicted output

**Example:**

$$\text{price} = 50000 + 200(\text{size}) + 10000(\text{rooms})$$

- Every extra square foot adds \$200
- Every extra room adds \$10,000

---

## 2. Why Add the Fake Feature ($x_0 = 1$)?

We rewrite the model by defining $x_0 = 1$, so:

$$\hat{y} = \sum_{j=0}^{p} w_j x_j$$

This is just **mathematical convenience** — the bias/intercept $w_0$ becomes part of the same unified formula, making matrix algebra much cleaner:

$$\underbrace{1 \cdot w_0}_{\text{bias}} + x_1 w_1 + x_2 w_2$$

---

## 3. Matrix Form

For many observations, we stack everything into matrices.

| size | rooms | price   |
|------|-------|---------|
| 1000 | 2     | 200,000 |
| 1500 | 3     | 300,000 |
| 2000 | 4     | 400,000 |

After adding the fake column of 1s, we get the **design matrix** $X$:

$$X = \begin{bmatrix} 1 & 1000 & 2 \\ 1 & 1500 & 3 \\ 1 & 2000 & 4 \end{bmatrix}, \quad w = \begin{bmatrix} w_0 \\ w_1 \\ w_2 \end{bmatrix}, \quad y = \begin{bmatrix} 200000 \\ 300000 \\ 400000 \end{bmatrix}$$

The model becomes the **central equation of linear regression**:

$$\boxed{y = Xw + \varepsilon}$$

---

## 4. The Error Term $\varepsilon$

Real data is noisy. Even two houses with identical features may differ in price due to location, construction quality, or hidden factors.

$$\text{actual value} = \underbrace{Xw}_{\text{systematic linear part}} + \underbrace{\varepsilon}_{\text{random noise}}$$

---

## 5. Gauss-Markov Assumptions

These assumptions are what make OLS mathematically "good".

### Assumption 1: Zero Mean Errors
$$E[\varepsilon] = 0$$
Errors average out to zero — the model is not systematically over- or under-predicting.

### Assumption 2: Homoscedasticity (Equal Variance)
$$\text{Var}(\varepsilon_i) = \sigma^2$$
The noise level is roughly constant everywhere. Violations occur when, e.g., expensive houses have much larger errors than cheap ones.

### Assumption 3: Uncorrelated Errors
$$\text{Cov}(\varepsilon_i, \varepsilon_j) = 0 \quad \text{for } i \neq j$$
One error should not influence another — especially important in time series or sequential data.

---

## 6 & 7. Ordinary Least Squares (OLS)

We don't know the true weights $w$ — we only observe data. So we compute an **estimate** $\hat{w}$.

**OLS chooses weights that minimize the sum of squared prediction errors:**

$$\min_w \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

### Why square the errors?
1. Positive and negative errors don't cancel each other out
2. Large errors are penalized more heavily
3. The math becomes analytically solvable

---

## 8. Geometric Intuition

OLS finds the line/plane that gets as close as possible to **all data points simultaneously**:

| # Features | Geometry        |
|------------|-----------------|
| 1          | Best-fit line   |
| 2          | Best-fit plane  |
| $p$        | Best-fit hyperplane |

> **Core intuition:** Minimize total squared *vertical* distances from each data point to the fitted surface.

---

## 9. The Normal Equation (Closed-Form Solution)

Minimizing the least-squares loss via calculus gives the **normal equation**:

$$\boxed{\hat{w} = (X^TX)^{-1}X^Ty}$$

### Breaking it apart intuitively:

| Term | Meaning |
|------|---------|
| $X^TX$ | Relationships (correlations) between features |
| $(X^TX)^{-1}$ | Corrects for overlap/redundancy among features |
| $X^Ty$ | Relationships between features and targets |

Combining them gives the weight vector that **minimizes squared error**.

---

## 10 & 11. Why OLS is Special — The Gauss-Markov Theorem

An estimator is called **linear** if it can be written as:
$$\hat{w} = Ay$$
for some matrix $A$. OLS satisfies this since:
$$\hat{w} = \underbrace{(X^TX)^{-1}X^T}_{A} \, y$$

The **Gauss-Markov theorem** states:

> *Among all linear unbiased estimators, OLS has the minimum variance.*

This makes OLS the **BLUE** estimator:

| Letter | Meaning |
|--------|---------|
| **B** | Best (lowest variance) |
| **L** | Linear (in $y$) |
| **U** | Unbiased (correct on average) |
| **E** | Estimator |

---

## Summary — The Full Picture

$$\text{Data} \rightarrow X, y$$

$$\text{Model} \rightarrow y = Xw + \varepsilon$$

$$\text{Goal} \rightarrow \text{find } w$$

$$\text{Method} \rightarrow \text{minimize squared errors}$$

$$\text{Solution} \rightarrow \hat{w} = (X^TX)^{-1}X^Ty$$

$$\text{Theory} \rightarrow \text{Gauss-Markov says OLS is optimal}$$

---

**The one-line intuition:**
> *Linear regression finds the hyperplane that minimizes total squared vertical distances to the data points.*

## 1. The Linear Regression Model

Suppose we want to predict a target variable $y$ from features $x_1, x_2, \dots, x_d$.

The linear model assumes:

$$y = w_0 + w_1x_1 + w_2x_2 + \dots + w_dx_d$$

where:

- $w_0$ = intercept (bias)
- $w_1,\dots,w_d$ = feature weights
- $d$ = number of features

This says:

> prediction = weighted sum of features

---

### Compact Vector Form

We usually add a fake feature $x_0 = 1$ so the intercept becomes part of the sum.

Then:

$$y = \sum_{j=0}^{d} w_j x_j$$

or in vector notation:

$$y = \mathbf{x}^T \mathbf{w}$$

where:

- $\mathbf{x} = (1, x_1, x_2, \dots, x_d)^T$
- $\mathbf{w} = (w_0, w_1, \dots, w_d)^T$

---

## 2. Dataset Form (Matrix Form)

Suppose we have $n$ observations.

We stack all observations into a matrix:

$$X = \begin{bmatrix} 1 & x_{11} & x_{12} & \dots & x_{1d} \ 1 & x_{21} & x_{22} & \dots & x_{2d} \ \vdots & \vdots & \vdots & & \vdots \ 1 & x_{n1} & x_{n2} & \dots & x_{nd} \end{bmatrix}$$

- rows = observations
- columns = features
- first column = all ones (bias term)

Target vector:

$$\mathbf{y} = \begin{bmatrix} y_1 \ y_2 \ \vdots \ y_n \end{bmatrix}$$

Weight vector:

$$\mathbf{w} = \begin{bmatrix} w_0 \ w_1 \ \vdots \ w_d \end{bmatrix}$$

Then the whole regression model becomes:

$$\mathbf{y} = X\mathbf{w} + \boldsymbol{\varepsilon}$$

This is the central equation of linear regression.

---

## 3. What is $\varepsilon$? (Error / Noise)

$\boldsymbol{\varepsilon}$ represents the part we cannot explain.

Real-world data is noisy:

- missing variables
- measurement error
- randomness
- imperfect model

So actual target values are:

$$y_i = \mathbf{x}_i^T \mathbf{w} + \varepsilon_i$$

where:

- $\mathbf{x}_i^T \mathbf{w}$ = model prediction
- $\varepsilon_i$ = random error

---

## 4. Gauss–Markov Assumptions

OLS depends on assumptions about errors.

---

### Assumption 1: Zero Mean Errors

$$E[\varepsilon_i] = 0$$

Meaning:

> errors average out to zero

The model is not systematically biased upward or downward.

---

### Assumption 2: Equal Variance (Homoscedasticity)

$$\text{Var}(\varepsilon_i) = \sigma^2$$

All errors have the same variance.

Graphically:

- spread of errors stays constant
- does not grow/shrink with $x$

---

### Assumption 3: Uncorrelated Errors

$$\text{Cov}(\varepsilon_i, \varepsilon_j) = 0 \quad (i \neq j)$$

Errors for different observations are unrelated.

---

## 5. What Does OLS Do?

OLS chooses weights that minimize prediction error.

Prediction:

$$\hat{\mathbf{y}} = X\mathbf{w}$$

Residuals (prediction errors):

$$\mathbf{y} - X\mathbf{w}$$

OLS minimizes the squared residuals:

$$L(\mathbf{w}) = \sum_{i=1}^{n} (y_i - \mathbf{x}_i^T\mathbf{w})^2$$

Matrix form:

$$L(\mathbf{w}) = (\mathbf{y} - X\mathbf{w})^T (\mathbf{y} - X\mathbf{w})$$

This is the Mean Squared Error objective.

---

## 6. Why Square the Errors?

If we used raw errors:

$$y_i - \hat y_i$$

positive and negative errors cancel.

Squaring:

$$(y_i - \hat y_i)^2$$

ensures:

- all errors are positive
- large errors are penalized more heavily
- objective becomes differentiable and convex

---

## 7. Solving the Optimization

We minimize:

$$L(\mathbf{w}) = (\mathbf{y} - X\mathbf{w})^T (\mathbf{y} - X\mathbf{w})$$

Take derivative with respect to $\mathbf{w}$:

$$\frac{\partial L}{\partial \mathbf{w}} = -2X^T(\mathbf{y} - X\mathbf{w})$$

Set derivative to zero:

$$X^T(\mathbf{y} - X\mathbf{w}) = 0$$

Expand:

$$X^T\mathbf{y} - X^TX\mathbf{w} = 0$$

Rearrange:

$$X^TX\mathbf{w} = X^T\mathbf{y}$$

Assuming $X^TX$ is invertible:

$$\boxed{\mathbf{w} = (X^TX)^{-1}X^T\mathbf{y}}$$

This is the famous **Normal Equation**.

---

## 8. Why Must $X$ Have Full Column Rank?

The article mentions:

$$\text{rank}(X) = d+1$$

This means columns of $X$ are linearly independent.

If not:

- some feature can be expressed using others
- $X^TX$ becomes non-invertible
- no unique OLS solution exists

This is multicollinearity.

---

## 9. What Does "Linear Estimator" Mean?

Estimator $\hat{\mathbf{w}}$ is called linear if it can be written as:

$$\hat{\mathbf{w}} = A\mathbf{y}$$

for some matrix $A$ depending only on $X$.

OLS satisfies this because:

$$\hat{\mathbf{w}} = (X^TX)^{-1}X^T\mathbf{y}$$

Here:

$$A = (X^TX)^{-1}X^T$$

depends only on $X$, not on $y$.

---

## 10. What Does "Unbiased Estimator" Mean?

An estimator is unbiased if:

$$E[\hat{\mathbf{w}}] = \mathbf{w}$$

Meaning:

> on average, it recovers the true parameters

OLS is unbiased under Gauss–Markov assumptions.

---

## 11. Gauss–Markov Theorem (Very Important)

The theorem says:

Among all **linear unbiased estimators**, OLS has the **minimum variance**.

This is called:

$$\textbf{BLUE}$$

**B**est **L**inear **U**nbiased **E**stimator.

Meaning:

- **Linear** → estimator linear in $y$
- **Unbiased** → correct on average
- **Best** → smallest variance

So OLS is the most stable linear unbiased estimator.

---

## 12. Geometric Interpretation

OLS projects $\mathbf{y}$ onto the column space of $X$.

The prediction:

$$\hat{\mathbf{y}} = X\hat{\mathbf{w}}$$

is the orthogonal projection of $y$.

Residuals:

$$\mathbf{r} = \mathbf{y} - \hat{\mathbf{y}}$$

are orthogonal to all columns of $X$:

$$X^T(\mathbf{y} - X\hat{\mathbf{w}})=0$$

This equation is exactly what we derived earlier.

---

## 13. Intuition Behind the Formula

The formula

$$\hat{\mathbf{w}}=(X^TX)^{-1}X^T\mathbf{y}$$

can be interpreted as:

- $X^T\mathbf{y}$ measures correlation between features and target
- $(X^TX)^{-1}$ corrects for feature overlap/interactions

Result:

- weights that best fit the data in least-squares sense

---

## 14. Connection to Maximum Likelihood

If errors are Gaussian:

$$\varepsilon_i \sim \mathcal{N}(0,\sigma^2)$$

then maximizing likelihood is equivalent to minimizing squared error.

That is why OLS is also a probabilistic estimator.

---

## 15. Key Big Picture

Linear regression is fundamentally:

> Find the hyperplane that minimizes squared distances to data points

OLS gives the exact analytical solution.

The assumptions guarantee:

- unbiasedness
- minimum variance
- statistical optimality

---

Loss:

$$L(\mathbf{w}) = (\mathbf{y}-X\mathbf{w})^T (\mathbf{y}-X\mathbf{w})$$

$$X^TX\mathbf{w}=X^T\mathbf{y}$$

OLS solution:

$$\boxed{\hat{\mathbf{w}} = (X^TX)^{-1}X^T\mathbf{y}}$$

> # Regularization of linear regression

YES. This is one of the most important ideas in all of machine learning:

# 🔥 Regularization = intentionally simplifying the model to make it stable.

This section is dense because it combines:

* linear algebra,
* geometry,
* statistics,
* optimization.

But the *core intuition* is actually simple.

---

# 🎯 THE BIG PROBLEM

OLS tries to find:

$$w=(X^TX)^{-1}X^Ty$$

But sometimes this becomes unstable or impossible.

Why?

Because:
$$X^TX$$
may be:

* singular,
* nearly singular,
* ill-conditioned.

---

# 🧠 Step 1 — What does "Singular Matrix" mean?

A singular matrix:

* has no inverse,
* behaves like division by zero.

---

# 🔥 Why does this happen?

Usually because features are highly correlated.

Example:

| Feature 1         | Feature 2            |
| ----------------- | -------------------- |
| area_with_balcony | area_without_balcony |

These contain almost the same information.

---

# 🎯 Multicollinearity

When features are strongly correlated:

$$x_1 \approx x_2$$

columns of $X$ become almost linearly dependent.

That makes:

$$X^TX$$

close to singular.

---

# 🧠 Why is this bad?

Because OLS uses:

$$(X^TX)^{-1}$$

If eigenvalues become tiny:

$$\lambda_i \to 0$$

then inverse eigenvalues become huge:

$$\frac{1}{\lambda_i} \to \infty$$

---

# 🔥 Intuition

Tiny changes in data →
massive changes in weights.

---

# 🎯 Example

Suppose model learns:

$$w_1 = 5000$$
$$w_2 = -4999$$

These huge weights almost cancel.

Add a few new samples:

Now:

$$w_1 = 10000$$
$$w_2 = -9998$$

Predictions might stay similar, but weights become unstable and meaningless.

---

# 🧠 This is called HIGH VARIANCE

The model becomes sensitive to small changes in training data.

---

# 🔥 Bias vs Variance Tradeoff

The article says:

> Sometimes we intentionally increase bias to reduce variance.

This is one of the deepest ML ideas.

---

# 🎯 High Variance

Model:

* unstable
* overfits
* changes drastically

---

# 🎯 Slight Bias

Model:

* simpler
* more stable
* generalizes better

---

# 🧠 Regularization solves this

Instead of minimizing ONLY prediction error:

$$||y-Xw||^2$$

we ALSO penalize large weights.

---

# 🔥 New Loss Function

$$\mathcal{L} = ||y-Xw||^2 + ||\Gamma w||^2$$

This second term is:

# 🎯 regularization penalty

---

# 🧠 Simplified Ridge Version

They choose:

$$\Gamma = \frac{\lambda}{2}I$$

which gives:

$$||w||_2^2$$

So final loss becomes:

$$||y-Xw||^2 + \lambda ||w||^2$$

This is:

# 🔥 Ridge Regression (L2 Regularization)

---

# 🎯 What does this mean intuitively?

Now the model wants TWO things:

---

## 1. Fit the data well

Small prediction error.

---

## 2. Keep weights small

Avoid huge unstable coefficients.

---

# 🔥 The Model Is Now "Conservative"

Without regularization:

```
"Use any crazy weights necessary to fit data."
```

With ridge:

```
"Fit data, BUT don't use ridiculously large weights."
```

---

# 🧠 Why does this stabilize the matrix?

Ridge changes:

$$X^TX$$

into:

$$X^TX+\lambda I$$

---

# 🎯 Key Insight

Adding:

$$\lambda I$$

adds positive values to diagonal entries.

This pushes tiny eigenvalues away from zero.

---

# 🔥 So the matrix becomes invertible and stable.

That's the REAL mathematical reason ridge works.

---

# 🧠 Final Ridge Solution

$$w = (X^TX+\lambda I)^{-1}X^Ty$$

Compare with OLS:

$$w=(X^TX)^{-1}X^Ty$$

Ridge simply adds:

$$+\lambda I$$

---

# 🎯 What does λ (lambda) control?

This is VERY important.

---

# Small λ

Behaves like OLS.

* low bias
* high variance

---

# Large λ

Strong regularization.

* higher bias
* lower variance
* weights shrink toward zero

---

# 🧠 Why do weights shrink toward zero?

Because ridge minimizes:

$$||w||^2$$

The easiest way to reduce that is:

$$w_i \to 0$$

---

# 🔥 Important Intuition

Ridge says:

> "Large weights are suspicious."

Because huge weights often indicate:

* overfitting,
* multicollinearity,
* instability.

### ridge regression

# Ridge Regression

That statement is referring to this ridge regression formula:

$$\hat{\beta}_{ridge}=(X^TX+\lambda I)^{-1}X^Ty$$

The important part is:

$$[ X^TX + \lambda I ]$$

where:

* $(X^TX)$ is the feature correlation matrix
* $(I)$ is the identity matrix (diagonal matrix)
* $(\lambda)$ is a positive number

## 1. Why ordinary regression can fail

Ordinary least squares uses:

$$[ (X^TX)^{-1} ]$$

But matrix inversion only works if $(X^TX)$ is invertible (non-singular).

Problems occur when:

* features are highly correlated
* one feature is a combination of others
* there are more features than samples

Then $(X^TX)$ becomes:

* singular
* or nearly singular

meaning:

* determinant near zero
* inverse unstable or impossible

## 2. What "regular matrix" means

In this context, "regular matrix" means:
an invertible matrix

Another term is:

* non-singular matrix

So ridge regression modifies $(X^TX)$ to make inversion possible and stable.

## 3. What is the "ridge"?

The "ridge" refers to adding:

$$\lambda I$$

to $(X^TX)$.

The identity matrix looks like:

$$I = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{bmatrix}$$

Multiplying by $(\lambda)$:

$$\lambda I = \begin{bmatrix} \lambda & 0 & 0 \\ 0 & \lambda & 0 \\ 0 & 0 & \lambda \end{bmatrix}$$

So ridge regression adds $(\lambda)$ to every diagonal element of $(X^TX)$.

## 4. Why adding to the diagonal helps

Suppose $(X^TX)$ has eigenvalues:

$$\lambda_1, \lambda_2, \dots$$

After adding $(\lambda I)$, they become:

$$\lambda_1+\lambda,\; \lambda_2+\lambda,\; \dots$$

This shifts all eigenvalues upward.

If some eigenvalues were:

* zero
* or extremely tiny

they become safely positive.

That guarantees invertibility.

## 5. Example

Suppose:

$$X^TX = \begin{bmatrix} 1 & 0.999 \\ 0.999 & 0.998 \end{bmatrix}$$

This matrix is nearly singular because the columns are almost linearly dependent.

Adding ridge term:

$$X^TX + 0.1I = \begin{bmatrix} 1.1 & 0.999 \\ 0.999 & 1.098 \end{bmatrix}$$

Now the matrix is much more stable to invert.

## 6. Geometric intuition

Without ridge:

* the optimization surface can be very flat in some directions
* tiny changes in data create huge coefficient swings

Adding $(\lambda I)$:

* "rounds out" the geometry
* stabilizes the solution
* prevents coefficients from exploding

## 7. Why it is called "ridge"

Historically, the name came from:

* ridge-like stabilization in the optimization landscape
* adding a ridge along the diagonal of the matrix

So "ridge regression" literally refers to:

* diagonal loading
* diagonal regularization
* adding $(\lambda I)$

## 8. Core idea in one sentence

Ridge regression works by adding a positive value to the diagonal of $(X^TX)$, making the matrix invertible and numerically stable while shrinking coefficients to reduce overfitting.
