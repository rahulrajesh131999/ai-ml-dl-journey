## Stemming and lemmatization

**Stemming** and **lemmatization** are two common **text preprocessing techniques** in Natural Language Processing (NLP). Both reduce words to a base form, but they do so differently.

| Feature          | Stemming                                     | Lemmatization                                   |
| ---------------- | -------------------------------------------- | ----------------------------------------------- |
| Definition       | Removes prefixes/suffixes using simple rules | Converts words to their dictionary (lemma) form |
| Uses dictionary? | No                                           | Yes                                             |
| Accuracy         | Lower                                        | Higher                                          |
| Speed            | Faster                                       | Slower                                          |
| Output           | May not be a real word                       | Always (or almost always) a valid word          |

## 1. Stemming

Stemming reduces words by chopping off prefixes or suffixes based on predefined rules. It does **not** consider the word's meaning or context.

### Example

| Original Word | Stemmed Word |
| ------------- | ------------ |
| playing       | play         |
| played        | play         |
| plays         | play         |
| studies       | studi        |
| running       | run          |
| happiness     | happi        |

Notice that:

* **studies → studi**
* **happiness → happi**

These are **not valid English words**, but they serve as a common root for matching similar words.

### Popular Stemming Algorithms

* Porter Stemmer
* Snowball Stemmer
* Lancaster Stemmer

### Advantages

* Fast
* Simple
* Uses less computation

### Disadvantages

* Can produce incorrect or meaningless stems
* Ignores grammar and context

---

## 2. Lemmatization

Lemmatization converts words to their **dictionary form (lemma)** by considering:

* vocabulary
* grammar
* part of speech (POS)

### Example

| Original Word | Lemma |
| ------------- | ----- |
| running       | run   |
| ran           | run   |
| better        | good  |
| studies       | study |
| mice          | mouse |
| am            | be    |

Unlike stemming, these outputs are valid dictionary words.

### Example with POS

Word: **meeting**

* Noun → meeting
* Verb → meet

The correct lemma depends on how the word is used.

### Advantages

* More accurate
* Produces meaningful words
* Better for applications requiring language understanding

### Disadvantages

* Slower
* Requires a dictionary and often POS tagging

---

## Example Comparison

Sentence:

> "The children are running and playing in the gardens."

| Word     | Stem     | Lemma  |
| -------- | -------- | ------ |
| children | children | child  |
| running  | run      | run    |
| playing  | play     | play   |
| gardens  | garden   | garden |
| are      | are      | be     |

Notice that:

* Stemming cannot reduce **children** to **child**.
* Lemmatization correctly converts **children → child** and **are → be**.

---

## Python Example

### Stemming (NLTK)

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "studies", "running", "happiness"]

for word in words:
    print(word, "->", stemmer.stem(word))
```

**Output**

```
playing -> play
studies -> studi
running -> run
happiness -> happi
```

---

### Lemmatization (NLTK)

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["playing", "studies", "running", "mice"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

**Output (without specifying POS)**

```
playing -> playing
studies -> study
running -> running
mice -> mouse
```

To get better results for verbs:

```python
lemmatizer.lemmatize("running", pos="v")
```

Output:

```
run
```

---

## When to Use Which?

### Use Stemming when:

* Speed is important.
* Building search engines or information retrieval systems.
* Exact word form is less important than grouping related words.

### Use Lemmatization when:

* Building chatbots or virtual assistants.
* Performing sentiment analysis.
* Doing machine translation or question answering.
* Any task where preserving meaning is important.

---

## Summary

| Aspect        | Stemming                         | Lemmatization                              |
| ------------- | -------------------------------- | ------------------------------------------ |
| Method        | Rule-based suffix/prefix removal | Dictionary + linguistic analysis           |
| Output        | May not be a real word           | Valid dictionary word                      |
| Accuracy      | Lower                            | Higher                                     |
| Speed         | Faster                           | Slower                                     |
| Context aware | No                               | Yes                                        |
| Example       | studies → studi                  | studies → study                            |
| Best for      | Search, indexing                 | NLP tasks requiring semantic understanding |

**In short:** Stemming is a quick, rule-based approach that trims words to a common root, while lemmatization is a linguistically informed approach that returns the correct dictionary form of a word.

# CountVectorizer unigram and bygram

This is one of those terms that sounds more complicated than it is. Let's build it up step by step.

### Step 1: What is a "gram"?

A **gram** is simply a sequence of words.

* **1 word** → **unigram**
* **2 words** → **bigram**
* **3 words** → **trigram**
* **n words** → **n-gram**

For example, take the sentence:

```text
I love machine learning
```

### Unigrams (1 word each)

```text
I
love
machine
learning
```

or as a list:

```python
["I", "love", "machine", "learning"]
```

---

### Bigrams (2 consecutive words)

Take every pair of neighboring words:

```text
I love
love machine
machine learning
```

or

```python
["I love", "love machine", "machine learning"]
```

---

### Trigrams (3 consecutive words)

```text
I love machine
love machine learning
```

or

```python
["I love machine", "love machine learning"]
```

---

## Step 2: What does `ngram_range` mean?

When you create a `CountVectorizer`:

```python
vect = CountVectorizer(ngram_range=(1,1))
```

the tuple `(1,1)` means:

> **Only include 1-word sequences (unigrams).**

So your vocabulary might become

```text
["I", "love", "machine", "learning"]
```

---

### `ngram_range=(2,2)`

Now only bigrams are used.

Vocabulary:

```text
["I love", "love machine", "machine learning"]
```

Notice that the individual words are **not** included.

---

### `ngram_range=(1,2)`

This is the most common setting.

It means:

> Include everything from **1-word sequences up to 2-word sequences**.

Vocabulary becomes:

```text
I
love
machine
learning
I love
love machine
machine learning
```

Now the model knows about both individual words **and** pairs of words.

---

## Why are bigrams useful?

Imagine these two sentences:

```text
I like this movie.
```

and

```text
I do not like this movie.
```

Using only unigrams:

Sentence 1:

```text
I, like, this, movie
```

Sentence 2:

```text
I, do, not, like, this, movie
```

The word **"like"** appears in both sentences.

But with bigrams:

Sentence 2 contains

```text
not like
```

This phrase has a very different meaning from just the word **"like"**. Bigrams help the model capture that.

---

## Another example

Sentence:

```text
The cat sat on the mat
```

### `(1,1)` → Unigrams

```text
The
cat
sat
on
the
mat
```

---

### `(2,2)` → Bigrams

```text
The cat
cat sat
sat on
on the
the mat
```

---

### `(1,2)` → Both

```text
The
cat
sat
on
the
mat
The cat
cat sat
sat on
on the
the mat
```

---

## In `CountVectorizer`

```python
from sklearn.feature_extraction.text import CountVectorizer

vect = CountVectorizer(ngram_range=(1,2))

X = vect.fit_transform(["I love machine learning"])
```

The learned vocabulary will look something like:

```python
{
    "i": 0,
    "love": 1,
    "machine": 2,
    "learning": 3,
    "i love": 4,
    "love machine": 5,
    "machine learning": 6
}
```

Notice that **"machine learning"** is treated as a single feature, not just two separate words.

---

### Quick summary

| `ngram_range` | What it includes                      |
| ------------- | ------------------------------------- |
| `(1,1)`       | Unigrams only (single words)          |
| `(2,2)`       | Bigrams only (pairs of words)         |
| `(3,3)`       | Trigrams only (groups of three words) |
| `(1,2)`       | Unigrams and bigrams                  |
| `(1,3)`       | Unigrams, bigrams, and trigrams       |

So in your code:

```python
vect = CountVectorizer(ngram_range=(1,1))
```

you're telling `CountVectorizer`:

> "Build my Bag of Words vocabulary using **only individual words**, not pairs or longer phrases."

This code measures **how similar names are** by representing each name as a vector of **character 3-grams** and then computing the **Euclidean distance** between those vectors.

Let's break it down.

---

## Step 1. Imports

```python
from scipy.spatial.distance import euclidean
from sklearn.feature_extraction.text import CountVectorizer
```

* `CountVectorizer` converts text into numeric vectors.
* `euclidean` computes the Euclidean distance between two vectors.

The Euclidean distance between vectors (x) and (y) is

[
\sqrt{\sum_i (x_i-y_i)^2}
]

* Smaller distance → more similar
* Larger distance → more different

---

## Step 2. Create the vectorizer

```python
vect = CountVectorizer(
    ngram_range=(3, 3),
    analyzer="char_wb"
)
```

### `ngram_range=(3,3)`

Use **only character trigrams** (3-character sequences).

Example:

```
cat
```

produces

```
cat
```

Example:

```
hello
```

produces

```
hel
ell
llo
```

---

### `analyzer="char_wb"`

Instead of splitting into words, it works with **characters**.

The `_wb` means **within word boundaries**.

It pads each word with spaces before extracting n-grams.

For example,

```
cat
```

becomes

```
" cat "
```

Then the trigrams are

```
" ca"
"cat"
"at "
```

Notice the spaces at the beginning and end.

This lets the model distinguish

```
cat
```

from

```
scatter
```

because `" ca"` only appears at the start of a word.

---

## Step 3. Fit and transform

```python
n1, n2, n3, n4 = vect.fit_transform(
    ["andersen", "petersen", "petrov", "smith"]
).toarray()
```

There are four names.

```
andersen
petersen
petrov
smith
```

### Build the vocabulary

The vectorizer first extracts every unique trigram.

For example,

### "andersen"

Pad:

```
" andersen "
```

Trigrams

```
 an
and
nde
der
ers
rse
sen
en
```

Actually with spaces they are

```
" an"
"and"
"nde"
"der"
"ers"
"rse"
"sen"
"en "
```

---

### "petersen"

Pad:

```
" petersen "
```

Trigrams

```
 pe
pet
ete
ter
ers
rse
sen
en
```

---

### "petrov"

Pad:

```
" petrov "
```

Trigrams

```
 pe
pet
etr
tro
rov
ov
```

More precisely

```
" pe"
"pet"
"etr"
"tro"
"rov"
"ov "
```

---

### "smith"

Pad:

```
" smith "
```

Trigrams

```
 sm
smi
mit
ith
th
```

Actually

```
" sm"
"smi"
"mit"
"ith"
"th "
```

---

All unique trigrams become the vocabulary.

---

## Step 4. Convert each name to a vector

Imagine the vocabulary were

| Index | Trigram |
| ----: | ------- |
|     0 | " an"   |
|     1 | " pe"   |
|     2 | "and"   |
|     3 | "pet"   |
|     4 | "ers"   |
|     5 | "sen"   |
|     6 | "smi"   |
|     7 | "ith"   |
|   ... | ...     |

Then

```
andersen
```

might become

```
[1,0,1,0,1,1,0,0,...]
```

while

```
petersen
```

becomes

```
[0,1,0,1,1,1,0,0,...]
```

Every vector has one component for each trigram in the vocabulary.

---

## Step 5. Compute distances

```python
euclidean(n1, n2)
```

compares

```
andersen
```

and

```
petersen
```

These names share

```
ers
rse
sen
en
```

So their vectors are fairly similar.

Distance is relatively small.

---

```python
euclidean(n2, n3)
```

compares

```
petersen
```

and

```
petrov
```

They share

```
 pe
pet
```

but differ after that.

Distance becomes larger.

---

```python
euclidean(n3, n4)
```

compares

```
petrov
```

and

```
smith
```

They share almost no trigrams.

Distance is much larger.

---

## Why character n-grams?

Character n-grams are useful for comparing words with similar spellings.

For example:

```
Andersen
Anderson
Andersson
```

share many trigrams, so their vectors are close.

Similarly,

```
John
Jon
Johnny
```

also share several character sequences.

This makes character n-grams useful for:

* detecting spelling mistakes,
* matching similar names,
* fuzzy search,
* record linkage (identifying duplicate records).
